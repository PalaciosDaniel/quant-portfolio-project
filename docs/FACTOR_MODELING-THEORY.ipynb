# Pipeline: Modelo de factores cross-sectional con ML + construcción de cartera

---
---

**¿Qué se entiendo por modelo en ciencia de datos?**

En ciencia de datos, un modelo es simplemente una función que aprende una relación entre unas entradas y una salida. En nuestr caso será algo como: tú le das de comer momentum, volatilidad, liquidez y tamaño y el modelo te devuelve el retorno esperado. 

**¿Qué son los factores?**

Un factor es una variable que creemos que contiene información útil para explicar o predecir los retornos de un activo. Originalmente, en los años 60 se desarrolló el modelo CAPM (Capital Asset Pricing Model) que decía que el rendimiento de una acción dependía de un solo factor: el riesgo de mercado $\beta$. Con el tiempo se han ido añadiendo más factores (hasta llegar a cientos de estos) pero los que están respaldados por décadas de datos académicos son: 

1. **Momentum (Impulso)**: se basa en la premisa de que lo que ha estado subiendo recientemente, tiende a seguir subiendo en el corto plazo, y lo que ha estado bajando, tiende a seguir bajando. Tiene sentido ya que los inversores suelen tardar en reaccionar a las buenas noticias (infravaloración inicial) y luego se suben al carro por miedo a quedarse fuera (efecto FOMO).

2. **Volatilidad**: mide cuánto varía el precio de una acción en un periodo de tiempo y se suele buscar la baja Volatilidad. Históricamente, las empresas más estables y defensivas suelen dar mejores rendimientos ajustados al riesgo a largo plazo que las empresas hipervolátiles.

3. **Tamaño (Size)**: este factor se centra en la capitalización bursátil de la empresa (el valor total de todas sus acciones). Generalmente se asocia al beneficio de invertir en Small Caps (empresas pequeñas) ya que las empresas pequeñas tienen más margen para crecer y expandirse que un gigante que ya domina el mundo.

4. **Liquidez (Liquidity)**: se refiere a la facilidad y rapidez con la que puedes comprar o vender un activo sin que su precio se vea drásticamente afectado. Los activos menos líquidos (más difíciles de vender rápido) suelen ofrecer una "prima de iliquidez", es decir, un rendimiento extra para compensar al inversor por el riesgo de quedarse "atrapado" en la posición.

5. **Value (Valor)**: consiste en buscar acciones que están "baratas" en comparación con sus fundamentos financieros reales (como el PER o el Precio/Valor). Es la base del famoso Value Investing (estilo Warren Buffett) y se basa en la lógica de que el mercado a veces se vuelve demasiado pesimista con algunas empresas sólidas, haciendo que su precio baje de forma injustificada. 

6. **Quality (Calidad)**: se enfoca en identificar empresas financieramente perfectas o muy robustas, con negocios estables y ventajas competitivas claras a través de métricas como los niveles de deuda, la rentabilidad sobre el capital (ROE), beneficios estables y buena gestión. Las empresas de alta calidad resisten mucho mejor las crisis económicas y generan flujos de caja constantes, lo que se traduce en un crecimiento sólido y seguro a largo plazo.

También existen otros factores como: 

- **Inversión (Investment / Asset Growth)**: Empresas que reinvierten de forma muy agresiva e imprudente suelen tener peor rendimiento que aquellas que son más conservadoras y eficientes con su capital.

- **Dividendo (Dividend Yield)**: El factor de buscar empresas que pagan dividendos altos y sostenibles de forma histórica (muy ligado a Value y Quality).

- **Factores Macroeconómicos**: Modelos como el APT (Arbitrage Pricing Theory) no miran características de la acción, sino factores externos como la inflación, los tipos de interés, el crecimiento del PIB o el precio de materias primas. 

> Destacar que no se utilizan todos siempre para predecir el retorno ya que estos factores no son predictivos a corto plazo sino que más bien te hablan de la tendencia histórica. 

**MODELO DE FACTORES**: un modelo que utiliza varios factores para intentar explicar o predecir el comportamiento futuro de las acciones.

**¿Qué significa cross-sectional?**

Existen dos formas principales de analizar datos financieros: 

1. **Time series (series tempoirales)**: se trata de coger un único activo y seguirlo a lo largo del tiempo. Aquí es donde entra la volatilidad y las correlaciones temporales. 

2. **Cross-sectional**: se trata de coger una fecha concreta y comparar unos activos con otros. Aquí entra la correlación entre activos y la diversificación. 

En nuestro caso nos decantamos más por un modelo cross-sectional porque para constuir una cartera no nos interesa tanto cómo de bien va a hacerlo un activo sino qué activo lo hará mejor con respecto a los demás, para ordenar sus ponderaciones.

**Machine Learning (ML)**

Digamos que el ML es el paso siguiente que nos aleja de la teoría financiera clásica y nos coloca a la vanguardia de las finanzas cuantitativas (Quantitative Finance). 
La diferencia entre el Aprendizaje Automático y las finanzas tradicionales es que antes el humano decidía la fórmula matemática concreta mientras que ahora el ML descubre por sí mismo a partir de los datos (rendimientos históricos y los 6 factores) la mejor fórmula para predecir el rendimiento futuro. Los cuatro algoritmos más comunes de más simples a los más avanzados son:

1. ***Regresión Lineal (El clásico)***: asume que la relación entre los factores y el rendimiento es una línea recta y lo que hace es intentar trazar la línea perfecta que pase lo más cerca posible de todos los puntos de datos históricos. El problema es que el mercado financiero casi nunca es lineal sino mas bien caótico, por lo que la regresión lineal a veces se queda muy "corta", es demasiado simple para capturar la tendencia o los patrones de los dato (underfitting).

2. ***Elastic Net***: es una regresión lineal pero con mecanismos de control de calidad (llamados regularización Lasso y Ridge) para evitar que el modelo se vuelva demasiado complejo  (overfitting) al intentar usar todos los factores. Lo que hace es penalizar al modelo si se vuelve demasiado complejo actuando como un filtro que apaga los factores que no sirven para nada (les pone peso cero) y encoge la importancia de los demás para que el modelo sea más robusto en el futuro.

3. ***XGBoost (Extreme Gradient Boosting)***: no usa líneas rectas sino Árboles de Decisión encadenados que se van constuyendo como una concatenación de decisiones que responden a preguntas. El "Gradient Boosting" significa que el algoritmo itera creando un nuevo árbol cada vez que se equivoca. Es uno de los algoritmos más ganadores en competiciones de ciencia de datos y es increíblemente preciso capturando relaciones raras y complejas del mercado.

4. ***LightGBM***: también es un modelo de árboles de decisión pero está optimizado para ser ultrarrápido y consumir menos memoria. Mientras que XGBoost analiza los árboles nivel por nivel (horizontalmente), LightGBM crece los árboles por hojas (verticalmente), yendo directo a donde puede reducir más el error.


Este es el primer paso del proyecto pero no el último. Una vez tengamos el score de cada uno de nuestros activos (que es lo que nos va a dar el modelo de factores) tenemos que constuir la cartera. Es decir, podemos saber que un activo es mejor que otro pero no sabemos qué pesos otorgar a cual, cuántos activos incluir, etc.

## Filosofía del proyecto

Antes de entrar en fases: lo que diferencia este proyecto de un "tutorial de Kaggle con acciones" es el rigor en tres puntos que un entrevistador quant va a mirar con lupa: (1) que la validación respete el orden temporal y no haya fuga de información del futuro, (2) que el backtest sea realista (costes de transacción, turnover, sin survivorship bias no documentado), y (3) que las conclusiones sean honestas incluso cuando el modelo no funciona tan bien como esperabais. Un Sharpe espectacular sin estas precauciones es una bandera roja para cualquiera que sepa de esto; un Sharpe modesto pero bien justificado metodológicamente vale mucho más en una entrevista.

Duración estimada total: 8-10 semanas a ritmo de verano (no full-time). Está pensado para que la complejidad crezca de forma incremental, así que si el verano se complica con otras cosas, el proyecto sigue siendo presentable parándolo en cualquier fase a partir de la 6.

**Las 3 claves del proyecto**

Si nos tenemos que quedar con algo sobre el proyecto dentro de un año, antes que quedarnos con el algoritmo de ML deberíamos de tener muy claro estos tres conceptos: 

1) **Que la validación respete el orden temporal y no haya fuga de información del futuro**: Este concepto se llama *look-ahead bias* o *data leakage* y se refiere a que es super importante diferenciar los datos que nuestro algoritmo utiliza para entrenar y los datos que utilizamos para validarlo. Es decir, con los datos del pasado predices el futuro (lo único es que para "saber" el futuro nos dejamos unos años sin utilizar para ponerlos como "futuro"). Esto se hace porque  el modelo termina aprendiendo cosas del futuro sin que seamos conscientes. 

> También podemos tener este *look-ahead bias* en la estructura que define la orden de compra. Si nuestro modelo calcula los factores utilizando el precio de cierre del lunes, esa orden de compra no se puede ejecutar con ese precio de cierre ya que por la noche los mercados estarían cerrados y realmente estarías comprando el martes por la mañana al precio de apertura del martes, que puede ser diferente. Para resolver este problema lo que haríamos es introducir un desfase entre la señal de compra y la compra en sí: la señal se genera con el precio de cierre del lunes y la compra con el precio de apertura del martes (si es mucho lío también podríamos esperarnos a comprar al precio de cierre del martes). 

2) **Costes de transacción por turnover (rotación de cartera)**: estos costes estarían implícitos en cada compra y venta de cada activo y es importante tenerlos en cuenta en la rentabilidad final. Tenemos varios costes de transacción: 

    - *Comisiones*: es lo que te cobra el bróker por comprar o vender y puede ser una tarifa fija por operación o un porcentaje del total. 

    - *El Spread o Horquilla*: las acciones no tienen un único precio, más bien tienen el precio al que puedes vender (bid) y el precio al que puedes comprar (ask). Obviamente el Ask siempre será más caaro que el Bid y la diferencia será el spread. Si comprar y vendes una misma acción en pocos segundos implícitamente te estás comiendo esta diferencia en cada transacción. 

    - *Impacto de Mercado*: si operas con grandes cantidades de dinero (o inviertes cantidades considerables en empresas pequeñas) puede pasar que el propio acto de lanzar una orden de compra repercuta en empujar el precio hacia arriba ya que los flujos que manejas son tan grandes que alteran la oferta y la demanda. 

    - *Impuestos*: en casi todo el mundo, cada vez que vendes una acción con ganancias, tienes que pagar un porcentaje al gobiernoagar. Esto es muy importante tenerlo en cuenta ya que las rotaciones de cartera deben de añadir suficiente rendimiento exrtra como para que compense pagar impuestos 

3) **Survivorship bias (sesgo de superviviencia)**: el sesgo de supervivencia es un error lógico que ocurre cuando haces un análisis centrándote solo en los sujetos que "sobrevivieron" a un proceso de selección, ignorando por completo a los que cayeron en el camino porque ya no son visibles. Esto puede darse cuando creamos muchas versiones diferentes de un mismo modelo variando ciertos parámetros aleatoriamente. Si todas las veriones fueron mediocres menos una que fue espectacular y nosotros borramos todas las vías menos esa no estamos mostrando que el algoritmo funciona sino más bien que hemos tenido suerte (estás forzando un resultado que saldría una de cada mil veces, como una especie de ad hoc donde rediriges hacia la dirección que tú quieres porque ya conoces los activos que han funcionado y los que no). En el caso de los datos se le conoce a esta práctica como *Data Snooping* o *Data Mining Bias*.

> También puedes tener este **survivorship bias** con las acciones. Si entrenamos el modelo con las acciones que componen un ínidce hoy en día (por ejemplo el SP500, que aglutinan las 500 mejores empresas de hoy) y miramos el rendimiento hacia atrás el modelo parecerá una mina de oro porque está dejando de lado todas las empresas gigantes que hace años quebraron y solo se está quedando con las que sobrevivieron. Estamos entrenando el modelo con información sesgada del futuro.

**MINI PIPELINE PARA ENTENDERLO TODO**

**OBJETIVO**: imaginemso que tebenis 500 empresas del universo que hayamos elegido. Nosotros simplemente queremos saber de la manera más riguirosa posible qué empresas creemos que serán las mejores en el próximo mes. Ahora vamos a ver las diferentes fases: 

1. **Conseguir los datos**: tenemos que obtener información de todas las empresas como precios diarios, volumen, capitalización, etc. 

2. **Construir los factores**: transformamos los datos en información útil como momentum, volatilidad, liquidez, tamaño, etc. 

3. **Crear la respuesta correcta**: necesitamos enseñarle al algoritmo qué queremos que aprenda, es decir, dados estos factores como entrada fijar una salida: retorno futuro. 

4. **Entrenar el modelo**: nutrimos al algortimo con miles de casos donde nosotros le mostramos la entrada y la salida para que empiece a crear patrones. 

5. **El modelo empieza a predecir**: una vez entrenado, probamos qué salidas futuras predice para unos factores actuales. 

6. **Convertir esas notas en una cartera**: un vez tenemos el ranking, construimos el portfolio de acciones. 

7. **Simular lo que habría ocurrido**: utilizamos esta estrategia en años pasados para ver qué tal lo hace. 

8. **Medir si realmente funciona**: calculamos métricas como el Sharpe, Drowdown, Volatilidad, CAGR, Sortino, Calmar, etc. 

9. **Validación del resultado**: debemos de asegurarnos que no hay fuga de información, si hay factor suerte, etc.

## Fase 0 — Definición de alcance y setup (3-4 días)

**Decisiones que hay que tomar antes de escribir código:**

- **Universo de activos**: empezad con algo manejable, por ejemplo los componentes actuales del S&P 500 o del Stoxx 600 (más fácil conseguir datos limpios de EEUU). Documentad desde el día 1 que estáis usando la composición *actual* del índice, no la histórica — esto introduce survivorship bias (las empresas que quebraron o fueron excluidas no aparecen). No pasa nada por tenerlo, pero hay que decirlo explícitamente en el README. Es mejor reconocer la limitación que pretender que no existe.
- **Frecuencia de rebalanceo**: mensual es el estándar académico para este tipo de modelos (datos diarios tienen demasiado ruido para factores fundamentales, datos anuales no dan suficientes observaciones). Empezad ahí.
- **Horizonte de predicción**: retorno a 1 mes vista (forward return), alineado con la frecuencia de rebalanceo.
- **Periodo histórico**: con yfinance podéis conseguir con garantías unos 10-15 años de precios diarios para casi cualquier large-cap. Apuntad a 2010-2024 o similar, dejando los últimos 1-2 años como holdout final que ninguno de los dos toca hasta el final del proyecto.

---
---

**¿Por qué la gran mayoría utilizaría la composición actual del índice?**

Sabemos que si decidimos utilizar la composición actual estamos introduciendo un survivorship bias porque solo nos estamos quedando con las empresas que han aguantado durante estos años en el índice, todas las que han quebrado o han sido expulsadas no entran en juego. Sin embargo, existen varias razones para considerar esta decisión: 

- **Dificultad de elegir la composición histórica**: tendríamos que conocer para cada fecha las empresas que se encotraban en el ínidce, las que salieron y las que entraron en cada momento, etc. y esos datos rara vez son gratuitos y limpios. 

- **Supervivencia no significa buena rentabilidad**: hay muhcas empresas que aunque han estado dentro del índice muchos años las rentabilidades que presentan no son buenas por lo que perdurar no siempre significa hacerlo bien. 

- **El objetivo nunca fue ganar dinero**: nosotros no queremos constuir una estrategia gnadora sino más bien entender el proceso de elegir los factores, entrenar los modelos, validarlos correctamente y constuir carteras. Vale más un Sharpe normalito pero bien explicado que uno espectacular producto de la suerte.   

> A pesar de los puntos que mencionamos como justificación, eso no significa que el sesgo desaparezca por lo que siempre vamos a tener en cuenta que este factor sobreestima ligeramente el rendimiento que podemos alcanzar con el modelo. 

---

**¿Por qué rebalancear mensualmente?**

La frecuencia de rebalanceo te dice cada cuánto tiempo actualizas tu distribución de activos. Esto a su vez dicta cada cuánto se recalulan los factores y se realizan nuevas predicciones y qué horizonte intenta predecir el modelo. 

Si rebalanceamos diariamente tendríamos que estar operando continuamente aumentando demasiado los costes. Además ya dijimos que la mayoría de factores cambian lentamente en el tiempo por lo que tan a corto plazo prácticamente no notaríamos diferencia. Por otro lado si eligiésemos rebalancear anualmente nos estaríamos quedando cortos porque en un año te puedes comer resultados empresariales inesperados, crisis, tendencias nuevas, etc.

Un rebalanceo mensual es justo el punto medio entre un rebalanceo diario demasiado frecuente y uno anual que supondría demasiado tiempo. Así tendríamos nueva información para cada rebalanceo, evitaríamos costes excesivos y seguiríamos teniendo sufucientes observaciones para entrenar el modelo. 

El **horizonte de predicción**, es decir, la cantidad de tiempo hacia el futuro hasta la que tu modelo va a realizar la predicción es el mismo que la frecuencia de rebalanceo ya que, si cada mez actualizamos la cartera, cada mes se realizará una nueva predicción. 

---

**¿Cómo elegir el periodo histórico para entrenar al modelo?**

Lo que si tenemos claro es que debemos de dejar un par de años sin tocar para luego verificar cómo predice nuestro modelo en un periodo de tiempo en el que no ha trabajado nunca. El problema viene con cuántos años incluimos en el entrenamiento del modelo. Si nos quedamos cortos (por ejemplo dos años) tendríamos muy pocas observaciones y justo podría coincidir con un único tipo de mercado (una rally alcista de dos años es perfectamente posible) y si cogemos muchísimos años surgiría el problema de que el mercado de hace treinta años no se parece en nada al actual (difusión de la información, traiding algorítmico, regulación, tipos de interés, etc.). 

Con 15 años, si la frecuencia de rebalanceo es mensual e incluimos 500 empresas tendríamos cerca de 90000 observaciones. A pesar de que el criterio sea un poco arbitrario, este número de años se puede considerar como una decisión metodológica razonable para este tipo de proyectos.

> Destacar también que no vamos a tener datos de las 500 empresas siempre porque a lo mejor hace 10 año alguna no existía o no era suficientemente grande como para entrar en el índice. De hecho puede pasar que una empresa haya entrado ya pero al no tener histórico suficiente no puedes calular factores que necesitan un año (como el momentum) para obtenerse. Esto es completamente normal y se tendrá en cuenta una vez obtengamos el "ranking de empresas" cuando tengamos que constuir la cartera. En ese momento decidiremos los requisitos para que una empresa pueda entrar a formar parte de nuestro portfolio (puede ser, por ejemplo, tener datos de esta en el último año).

**¿Por qué es importante el set up técnico?**

El **set up técnico** simplemente es la manera en que vamos a ir organizando todo el código que vayamos construyendo. Este proyecto es suficientemente grande como para tener que dividir las diferentes partes en varios archivos. El **repositorio** es la carpeta principal del proyecto y dentro de él estará tanto el código como la documentación, las imágenes, etc. Lo bueno de hacerlo en Github es que funciona como una nube y permite que trabajen varias personas. 

El objetivo de **ordenar por carpetas** es que dentro de tres meses te puedas meter y entiendas dónde está cada cosa. La estructura obvia es carpetas con subcarpetas dentro y las que no pueden faltar son: 

- Carpeta ***data/*** : aquí metemos todos los datos como precios de activos, volúmenes, etc. Puedes tener subcarpetas donde tengas una para los datos crudos ***raw/*** tal cual los extraes, otra para los procesados ***processed/*** donde los dejas bien limpios y ordenados y otra para datos que aporten contexto ***external/*** donde metes diccionarios de activos (relación entre ticker y empresa), datos macro globales (datos de inflación, interes,etc.) o calendarios de mercado (días raros donde el mercado cierra). Dentro de estas carpetas nos vamos a encontrar archivos tipo *precios.csv*, *factores.csv*, etc.  

- Carpeta ***src/*** : aquí encontramos el código de verdad, donde encontraremos el programa. Dentro podemos tener el código que construye los factores ***features/*** que dentro a su vez puede contener los diferentes factores (momentum.py, volatility.py, etc), todo lo relacionado con entrenar los modelos ***models/*** o todo lo que ocurre después de obtener las predicciones ***backtest/*** donde encontramos la construcción del portfolio, los costes o los backtests. También podemos encontrar el código principal que se ejecuta para mostrar todos los resultados ***main.py*** y la carpeta ***utils/*** donde incluimos todos las funciones de soporte, cálculos matemáticos y herramientas repetitivas que ensuciarían el código prinicpal. 

- Carpeta ***notebooks/*** : serían Jupyter notebooks donde pruebas todas las ideas que quieras, graficas y calulas cosas pero no es él definitivo. Si quieres salvar algo de tu experimento lo pasas a .py y lo metes en la carpeta correspondiente. Los notebooks, si el proyecto es grande, acaban siendo difíciles de matener. 

- Carpeta ***test/*** : aquí incluimos todos los programas que tienen como objetivo comprobar que el resto del código funciona bien. 

- Carpeta ***configs/*** : aquí metemos todos los parámetros, nombres de variables, etc. para no tener que estar repitiéndolas todo el rato en cada código.

- Archivo de **README.md** es el documento de bienvenida que cualquiera ve en cuanto entra en el repositorio de GitHub. Aquín encontramos información esencial como el objetivo del proyecto, el procedimiento y estrcutra, las intrucciones de instalación, etc. 

**¿Qué es un entorno reproducible?**

Un entorno reproducible en programación es un sistema que garantiza que tu código funcione exactamente de la misma manera, produzca los mismos resultados y no dé errores, sin importar en qué ordenador se ejecute, quién lo corra o cuánto tiempo haya pasado desde que se escribió. Para crear este entorno debemos fijar las versiones de las librerías que vamos a utilizar y todo eso lo incluimos en la carpeta ***requirements.txt***. 

EL problema es que podemos tener modelos pertenecientes a librerías especializadas que dependen de otras librerías, por lo que, por mucho que fijemos la versión de la primera el código podría fallar por esa dependencia oculta. Para evitar esto nació ***poetry*** que sirve como una herramienta todo en uno para la gestión de entornos y dependencias en Python. En vez de tener nuestra *requirements.txt* tendríamos estos dos archivos: 

- Carpeta ***pyproject.toml*** : esta sería el sustituto de *requirements.txt*, donde configuramos todo. Para instalar nuevas librerías, en vez de añadirla, simplemente vas a la terminal y escribes "*poetry add scikit-learn*". Poetry analiza automáticamente todas las sub-librerías necesarias para correr esta y actualiza el archivo por ti. 

- Carpeta ***poetry.lock*** : sería como el candado con las versiones exactas, donde guardas la versión exacta y un código de seguridad de absolutamente cada micro-librería instalada. Esta carpeta es lo que utilizaría otra persona para reproducir el mismo entorno. 

A la hora de compartir el proyecto, la otra persona simplemente tendrá que bajarse poetry, instalas en la terminal *poetry install* y automáticamente poetry reconocerá el archivo *poetry.lock* e instalará exactamente la misma combinación de bits que teníamos nosotros cuando el modelo funcionaba.

**Cuidado con los CSVs pesados**

Github está pensado para almacenar código pero no enormes cantidades de datos. Por eso se utiliza ***.gitignore*** como una lista de cosas que Github debe ignorar como ***data/***, ssería como un portero de discoteca con una lista negra que filtra los archvios que se van a subir a internet y los que no. Cuando nosotros corramos el programa de pyhton, el script descargará los gigabytes de precios y los escribirá en *data/precios.csv*, esos archivos están físicamente ocupando espacio en el ordenador pero nunca viajarán a GitHub porque *.gitignore* los tiene bloqueados.  

Si otra persona se baja tu proyecto desde Github le aparecerá todo el código menos los archivos incluidos en el *.gitignore*. Esto no solo se hace por un tema de esapcio sino por eficiencia (es mucho más rápido descargar de internet que mover millones de filas y columnas de un lado a otro) y por derechos de autor o privacidad (las bases de datos de pago no se pueden subir a internet así como así). 

Las intrucciones de cómo descargarse estos datos aparecerán en el **README.md** donde especificas exactamente el código que debes correr para obtener todos los datos necesarios y crear los archivos donde estos se guarden.

## Fase 1 — Adquisición de datos (1-1.5 semanas)

**Precios:**

`yfinance` para precios diarios ajustados (ajustados por dividendos y splits, importante para que los retornos sean correctos). Aquí ya hay una decisión sutil: el ajuste por dividendos cambia los retornos históricos, así que aseguraos de que estáis usando `Adj Close` y no `Close`.

**Datos fundamentales (para factores tipo value, quality):**

Esto es lo más difícil de conseguir gratis con calidad point-in-time (es decir, el dato tal y como se conocía en su momento, sin revisiones posteriores). Las opciones razonables para un proyecto de estudiantes:

- `yfinance` también trae algunos fundamentales pero son del momento actual, no históricos — sirven solo si simplificáis el proyecto a un único snapshot transversal (menos interesante).
- Simfin (tiene tier gratuito) ofrece históricos fundamentales con más profundidad.
- Alternativa más simple y honesta: limitad los factores fundamentales a los que se pueden derivar de precios y volumen únicamente (momentum, volatilidad, tamaño vía capitalización aproximada, liquidez), y dejad fuera factores tipo P/E o P/B si no tenéis garantías de la calidad del dato histórico. Un proyecto con 5 factores bien construidos y documentados vale más que uno con 15 factores de calidad dudosa.

**Recomendación concreta**: empezad solo con factores derivables de precio/volumen (fase 2 los detalla). Si os queda tiempo en julio/agosto, añadís factores fundamentales como extensión. Así no os bloqueáis semanas enteras peleando con APIs de datos fundamentales gratuitas, que es donde mueren muchos proyectos de este tipo.

**Estructura de almacenamiento**: un panel (panel data) con multi-índice (fecha, ticker) es la estructura natural. Pandas con `MultiIndex` o, si os apetece algo más robusto, `polars` (más rápido, y mencionarlo en el CV no hace daño porque cada vez se usa más en la industria).

**División de trabajo**: tu compañero construye el pipeline de descarga y el almacenamiento (con manejo de errores, reintentos, caché para no golpear la API cada vez); tú validas que los datos tienen sentido financiero (¿hay splits raros sin ajustar? ¿hay gaps sospechosos? ¿los retornos diarios tienen magnitudes razonables?).

**¿Cómo funcionan los dividendos?**

Cada vez que compras una accción de una empresa y esta obtiene beneficios podría reinvertir ese dinero, comprar otras empresas, reducir deuda o repartir parte de esos beneficios a los accionistas. Esto último es lo que llamamos **dividendos**. Cuando una empresa reparte dividendos ese dinero sale físicamente de la cuenta bancaria de la empresa y va a parar a la cuenta de cada accionista (según el número de acciones que se tengan) por lo que ahora la empresa tendría menos efectivo disponible y el valor real de la empresa (su valor contable) baja. 

Para reflejar esta realidad, ocurre un ajuste automático. La noche anterior al día de reparto (un día clave llamado **Fecha Ex-Dividendo**), la propia sociedad que gestiona la Bolsa de Valores (por ejemplo, el New York Stock Exchange o BME en España) resta artificial y matemáticamente el dividendo por acción al precio de cierre de la acción por lo que a la mañana siguiente estaría cotizando a menor precio. Esto se hace porque sino tendrías una oportunidad de arbitraje donde ganarías dinero sin asumir ningún riesgo. 

**¿Cómo funciona el split?**

Un split o desdoblamiento de acciones es un ajuste financiero donde la empresa aumenta el número de acciones en circulación dividiendo las ya existentes, lo que reduce automáticamente el precio de estas. El split más común es el 2:1 (si antes una acción valía 100$ ahora tendrás dos acciones que valen 50$) pero puedes tener split 20:1 o incluso split inverso. 

Las razones para hacer un split pueden ser por accesibilidad (si la acción se vuelve muy cara habría mucha gente que no podría comprar permitirse ni una y no compraría) o por liquidez (al haber muchas más acciones en circulación y a un precio más bajo, es mucho más fácil y rápido que la gente las compre y las venda)


Diferencia entre ***Close*** y ***Adjusted Close*** : mientras que el primero simplemente es el precio de cierre de la acción cada día, en el *Adjusted Close* modifica el histórico de precios para que el cálculo de los retornos refleje la riqueza total del inversor, incluyendo el dividendo como parte de la rentabilidad y teniendo en cuenta los splits que hayan hecho. 

---

**¿Y por qué utilizar los precios de cierre?**

Cada acción diariamente tiene un Close (pecio de cierre) pero también un Open (primer precio del día), un High (precio máximo alcanzado) y un Low (precio mínimo alcanzado). Normalmente para constuir retornos, índices bursátiles, factores, volatilidad, etc. se utiliza el Close porque el precio al terminar toda la sesión diaria, como si fuese el resumen del día. 

**Concepto de *Overnight***

A pesar de que las bolsas cierren el mercado sigue activo "por la noche": las empresas publican resultados por la tarde, los presidentes o bancos centrales dan discursos de madrugada y los mercados en otras partes del mundo siguen abiertos. En finanzas cuantitativas modernas se suele separar el rendimiento overnight del rendimiento intradía ya que históricamente se ha demostrado que el comportamiento humano y de los algoritmos es completamente diferente de noche que de día. Incluso se ha visto que el "efecto Overnight" suele ser alcista mientras que el intradía más plano o volátil. 

> Destacar que si solo operamos con precios de cierre estamos incoporando el rendimiento Overnight. 

Tanto el High como el Low pueden darnos problemas porque estos precios se alcanzan durante unos pocos segundos con muy poquitas acciones, por lo que los valores que se manejan aquí no van a ser muy representativos del precio del día. También podríamos operar con una media diaria o algo parecido pero para ello necesitaríamos datos intradía y no tiene sentido complicar más el tema.

## Fase 2 — Feature engineering: construcción de factores (1.5-2 semanas)

Esta es la fase donde tu conocimiento de finanzas pesa más. Factores típicos, derivables solo de precio/volumen:

- **Momentum**: retorno acumulado de los últimos 12 meses excluyendo el último mes (el "12-1 momentum" clásico de Jegadeesh-Titman, excluir el último mes evita el efecto de reversión a corto plazo).
- **Volatilidad**: desviación estándar de retornos diarios en una ventana móvil (por ejemplo 60 o 90 días). Los activos de baja volatilidad históricamente tienen mejor retorno ajustado a riesgo del que predicería el CAPM — el famoso "low volatility anomaly", que es un buen tema para hablar en una entrevista si os preguntan por qué incluisteis ese factor.
- **Tamaño**: log de la capitalización de mercado (precio × acciones en circulación). El "size premium" es uno de los factores de Fama-French más estudiados.
- **Liquidez**: volumen medio o algo como el Amihud illiquidity ratio.
- **Reversión a corto plazo**: retorno del último mes (efecto contrario al momentum, a más corto plazo).
- (Opcional, si conseguís fundamentales fiables) **Value**: book-to-market o earnings yield.

**Puntos técnicos importantes:**

- **Normalización cross-sectional**: en cada fecha de rebalanceo, normalizad cada factor *entre los activos de ese momento* (z-score o rank percentil), no a lo largo del tiempo. Esto es lo que hace que el modelo aprenda "qué activos son relativamente mejores en este momento" en lugar de mezclar escalas de épocas distintas.
- **Winsorización**: los factores financieros tienen colas pesadas (esto lo sabes de tus modelos GARCH/Student-t con BTC), así que recortad outliers extremos (por ejemplo al percentil 1 y 99) antes de normalizar, o el z-score se distorsiona con un solo valor extremo.
- **Lag correcto**: si usáis datos fundamentales, hay que aplicar un lag de publicación realista (los resultados del Q4 no se conocen el 1 de enero, normalmente se publican semanas o meses después). Con factores de precio/volumen este problema es menor, pero hay que tenerlo en mente para cuando añadáis fundamentales.

**División de trabajo**: tú defines y justificas financieramente cada factor (por qué debería tener poder predictivo, qué anomalía de mercado captura); tu compañero implementa el cálculo vectorizado y eficiente (evitar loops sobre miles de combinaciones fecha-ticker, usar `groupby` + `transform` en pandas o las primitivas equivalentes en polars).

---
---

**¿Por qué se utilizan factores?**

Cada factor describe una manera de cuantificar una característica propia del activo. El modelo de Machine Learning no entiende de conceptos abstractos cualitativos con los que nosotros nos manejamos y que asociamos a las empresas y los factores son la forma de convertir esta empresa y todas sus cualidades en vectores de números con lo que el modelo sí sabe trabajar. 

---

#### **Momentum**

Rendimiento acumulado de una acción durante los últimos 12 meses exceptuando el último mes (por eso se llama **12-1 Momentum**) y mide cómo de ganadora ha sido una empresa en el último año. Se basa en la premisa (siendo históricamente uno de los factores más robustos encontrados en finanzas) de que las acciones que han ido bien durante bastante tiempo tienden, de media, a seguir haciéndolo bien durante un tiempo más. 

> Psicológicamente el **momentum** tiene mucho sentido ya que los inversores suelen tardar en reaccionar a las buenas noticias (infravaloración inicial) y luego se suben al carro por miedo a quedarse fuera (efecto FOMO).

**¿Por qué excluir el útlimo mes?**

El último mes se excluye para evitar la **reversión a corto plazo** (Short-term Mean Reversion) que describe el efecto contrario: acciones que han subido mucho este mes a menudo corrigen ligeramente el mes siguiente. Lógicamente a corto plazo el mercado se mueve por puro ruido, liquidez y flujos técnicos y si una acción sube mucho en poco tiempo los inversores querrán recoger beneficios y los creadores de mercado (market makers) ajustarán sus carteras. 

> Al eliminar el último mes nos quedamos con la tendencia a medio plazo y evitamos los movimientos muy recientes. 

**¿Por qué 12 meses?**

En 1993, dos investigadores llamados Narasimhan Jegadeesh y Sheridan Titman publicaron un estudio revolucionario donde probaron científicamente por primera vez el factor Momentum y sus conclusiones fueron que **el horizonte de entre 6 y 12 meses era el "punto dulce"** (sweet spot) donde el efecto Momentum era más fuerte y consistente. Desde ese momento la industria adoptó los 12 meses como el estándar oficial. 

> Podríamos pensar que el horizonte temporal que se maneja es un poco arbitrario y debería depender del momento histórico o sector donde nos encontremos. Sin embargo, no siempre especificar al máximo (por ejemplo cambiar el horizonte dependiendo del sector) es bueno ya que puedes caer fácilmente es un Overfitting (Sobreajuste): si nos ponemos a buscar la combinación perfecta por sector, acabaremos encontrando una combinación matemática que funcionó en el pasado por pura casualidad, pero que fracasará estrepitosamente en el backtest fuera de muestra o en el futuro.


---


#### **Reversión a corto plazo**

Justamente podemos utilizar el efecto que descartamos en el Momentum como otro nuevo factor independiente. En este caso trabajaríamos con los retornos de las últimas semanas (o como mucho del último mes). Aquí ya no estarías capturando la psicología del "rebaño" o las tendencias macroeconómicas sino los límites físicos de la liquidez y los excesos emocionales de corto plazo. 

> El hecho de diferenciar el momentum de la revsersión a corto plazo y utilizar ambas como factores independientes le enseña a nuestro modelo de ML a identificar las caídas fuertes momentaneas (con la oportunidad de comrpa que eso conlleva) dentro de una tendencia alcista a largo plazo.


---


#### **Volatilidad**

Desviación estándar de los retornos diarios durante una ventana móvil o periodo de tiempo. Se suele buscar la baja volatilidad ya que hsitóricamente las empresas menos volátiles han obtenido mejores retornos ajustados por riesgo de los que predeciría el modelo CAPM.  

> El CAPM es el modelo financiero clásico, que encaja dentro de la Teoría Moderna de Portafolio o MPT, que permite calcular la rentabilidad mínima que un inversor debe exigirle a un activo a cambio del riesgo que asume. 

**Low Volatility Anomaly**

El hecho de que las acciones con menor volatilidad (más estables y aburridas) obtengan mejores rendimientos ajustados al riesgo que las acciones hipervolátiles choca directamente con la teoría clásica (CAPM) ya que en finanzas clásicas se creía que una mayor rentabilidad implicaba sí o si un mayor riesgo. 

> Psicológicamente esto se explica por el "**Efecto Lotería**", los inversores minoristas prefieren comprar acciones baratas y ultravolátiles con la esperanza de dar el "pelotazo" lo cual hace que el precio se infle destruyendo la rentabilidad futura, y por el "**Sesgo de exceso de confianza**", los inversores tienden a sobreestimar su capacidad de salirse justo antes del desplome creyéndose más listos que los demás, lo cual hace que compren activos peligrosos en masa. 

> También existe un factor estrucutral debido a la **restricción de apalancamiento de los Fondos**. Al tener prohibido por ley apalancarse para comprar acciones, si un fondo quiere batir a su índice de referencia, la única opción que le queda para intentar ganar más es comprar las acciones de mayor riesgo lo que empuja hacia arriba el precio de las acciones volátiles.

**¿Y si separamos las subidas de las bajadas?**

Históricamente se ha comprobado que las caídas en bolsa suelen ser mucho más rápidas, violentas y agresivas que las subidas, que suelen ser más escalonadas. Mientras que las caídas están movidas por el pánico colectivo, la necesidad urgente de liquidez y la ejecución automática de órdenes de stop-loss, las subidas obedecen a un optimismo que gradualmente va calando en los inversores. 

> Aquí entra el **Leverage Effect** que explica que cuando una empresa cae, su estructura de deuda se vuelve más arriesgada, lo que desata aún más volatilidad hacia abajo.

En el intento de diferenciar dos volatilidades distintas surge en finanzas cuantitativas modernas la rama que llamamos "**Gestión del Riesgo Asimétrico**" (Downside Risk). La solución a este problema vino por diferenciar dos variables distintas: 

- Upside Volatility: mide la desviación estándar solo de los días en que la acción cierra en positivo. 

- Downside Volatility: mide la desviación estándar solo de los días en que la acción cierra en negativo. 

> Este segundo factor da origen al **Ratio de Sortino**, que sustituye al Ratio de Sharpe porque solo castiga al modelo por las caídas.


---


#### **Tamaño (Size)**

Este factor se centra en la capitalización bursátil de la empresa, que se calcula como el precio de una acción multiplicado por el número de acciones. Los datos necesarios para calcular el tamaño son el precio y el número de acciones en circulación. Generalmente se asocia al beneficio de invertir en Small Caps (empresas pequeñas) por encima de las grandes empresas. 

**Size premium** 

Esta prima por tamaño originalmente fue demostrada por el economista Rolf Banz en 1981, llegando a la conclusión de que las empresas pequeñas (Small-Caps) tienden a generar mayores rentabilidades a largo plazo que las empresas gigantes (Large-Caps). Esta Prima por tamaño contradice la teoría clásica que argumentaba que solo las empresas grandes y consolidadas son las mejores inversiones. Los principales factores que explican esta prima son: 

- **Mayor potencial de crecimiento**: las empresas pequeñas tienen mucho más margen para crecer y expandirse que un gigante que ya domina el mundo. 

- **La compensación por el Riesgo de Liquidez**: como las acciones de las empresas pequeñas se compran y se venden mucho menos, un fondo grande encontrará mayor dificultad si se quiere salir de golpe ya que no encontrará suficientes compradores y tendrá que tirar el precio. Esto convierte a las Small Caps en inversiones más incómodas y de mayor riesgo contractual y el mercado compensa a los inversores ofreciendo una rentabilidad media más alta a largo plazo. 

- **La falta de atención (Infiltración de información)**: las empresas grandes están siendo constantemente analizadas por lo que resulta imposible encontrar cualquier ineficiencia. Sin embargo, las pequeñas empresas al no estar en el punto de mira, permite que los inversores listos encuentren estas ineficiencias antes que el resto.

> Destacar que en los últimos 15-20 años, el Size Premium casi ha desaparecido en los mercados desarrollados. Las empresas gigantes tecnológicas han crecido tanto y de forma tan eficiente que han aplastado a las pequeñas. Muchos académicos debaten hoy en día si el factor Size sigue vivo o si fue una casualidad del siglo pasado.

> Destacar también, como otra posible derivada del proyecto, que el **factor size es el único aprovechable para los inversores minoristas** ya que los grandes fondos son demasiado grandes como para meterse en estos activos. Podríamos implementar el ML no solo en el SP500 sino en índices que aglutinen empresas de menor tamaño diferenciando así el rendimiento de nuestro algoritmo para Large-Caps respecto del de Small-Caps. 

---

#### **Liquidez (Liquidity)** 

Se refiere a la facilidad y rapidez con la que puedes comprar o vender un activo sin que su precio se vea drásticamente afectado. Los activos menos líquidos (más difíciles de vender rápido) suelen ofrecer una "prima de iliquidez", es decir, un rendimiento extra para compensar al inversor por el riesgo de quedarse "atrapado" en la posición. Saber cómo de líquido es un activo es crucial para asegurarte de que vas a poder salir y entrar en el momento y en la cantidad que te estimula el modelo. 

Para calcular esta liquidez tenemos dos opciones, la forma sencilla en la que aproximas la liquidez al volumen medio y otra más sofisticada que relaciona el movimiento del precio con el volumen negociado.

**Volumen medio de negociación**: nos referimos al volumen de acciones o volumen de efectivo que cambian de manos en un día, siendo el segundo el número de títulos individuales intercambiados multiplicados por el precio de la acción. 

**¿Por qué aproximar liquidez por el volumen?**

El volumen de negociación te está dando una métrica a groso modo sobre cuánto dinero se está intercambiando en un día, es decir, cuánto dinero disponible hay para comprar y vender y con eso te puedes hacer una idea de el grado de liquidez. Sin embargo, esta métrica solo te habla de cuánto se comercia pero no de qué pasa con el precio cuando se comercia (una acción puede tener mucho volumen un día simplemente porque se está desplomando y todo el mundo está vendiendo). 

Amihud Illiquidity Ratio: el economista Yakov Amihud creó en 2002 un ratio (actualmente el estándar en la industria) que conecta el volumen con el movimiento del precio, es decir, mide el impacto en el precio por cada dólar negociado (nos da una medida de cuánto dinero es necesario meter para generar un impacto significativo en el precio, lo cual es una medida perfecta de lo líquida que es esa empresa). 

$$\text{Amihud}_{t} = \frac{|\text{Rentabilidad de ese dia}|}{\text{Volumen en Efectivo de ese dia}}$$

Cuanto más alto sea el ratio menos líquida es la empresa. 

---

#### **Value, Quality, Inversión y Dividendos**

Para calcular estos cuatro factores necesitamos los **Estados Financieros** de las empresas. Principalmente tres documentos que las empresas públicas están obligadas a presentar cada trimestre: el Balance de Situación, la Cuenta de Resultados y el Estado de Flujos de Efectivo.

- **Balance de Situación (balance sheet)**: es como una fotografía instantánea de lo que la empresa tiene y debe en un momento exacto. Muestra la estructura financiera de la compañía bajo la premisa de que lo que la empresa tiene como activos se financia con deuda o con recursos propios siendo el patrimonio neto la diferencia entre los activos y los pasivos. 

- **La Cuenta de Resultados (Income Statement / P&L)**: muestra si la empresa ha ganado o perdido dinero en ese periodo de tiempo. Para ello se le restan a los Ingresos totales los Costes de producción, los Gastos operativos, los Intereses de la deuda y los Impuestos obteniendo así el **Beneficio Neto**.  

- **El Estado de Flujos de Efectivo (Cash Flow Statement)**: este estado financiero sigue el rastro del dinero en efectivo que entra y sale durante el trimestre. Se divide en actividades de Operación (dinero que entra y sale por el negocio principal), actividades de Inversión (dinero gastado o recibido por comprar o vender activos a largo plazo) y actividades de Financiación (dinero que se mueve por pedir préstamos, devolver deuda, emitir acciones o pagar dividendos a los inversores).

Para cada factor identificaremos en estos informes los datos espcíficos necesarios y crearemos las nuevas variables en cada caso:  

1. **Factor Value (Buscar empresas baratas)**: necesitamos el beneficio neto por accióm, las ventas totales o el valor contable de los activos para poder calcular el ratios como el **P/E** (Price-to-Earnings), **P/B** (Price-to-Book) o el **EV/EBITDA**.

2. **Factor Quality (Buscar empresas excelentes y estables)**: necesitamos el beneficio operativo, el capital invertido y la deuda total para poder calular el **ROE** (Return on Equity), el **ROIC**, o el margen de beneficio neto.

3. **Factor Inversión (Investment)**: necesitamos el **CapEx** (Capital Expenditures) y el crecimiento total de sus activos para poder calular el crecimiento anual de los activos o ratio CapEx/Ventas.
 
4. **Factor Dividendos (Dividend Yield)**: necesitamos el dinero en efectivo que la empresa paga por acción al año para poder calcular el **Dividend Yield**.

#### **Factores Macroeconómicos**

A diferencia de los anteriores, estos factores son los mismos para todas las acciones en un momento dado. Los datos que queremos en este caso son los Tipos de Interés fijados por los bancos centrales (como la Reserva Federal o el BCE), la tasa de Inflación (IPC), el crecimiento del PIB (Producto Interior Bruto) o el Índice VIX (el indicador del miedo del mercado).

> Así el modelo podrá diferenciar entre diferentes contextos y encontrar relaciones entre periodos con unas caracteríticas particulares y empresas con unos determinados factores que lo hicieron muy bien en esos momentos.

## **Procesado de factores**

Una vez obtenemos los factores que hemos elegido ahora debemos tranformarlos antes de dárselos al modelo para que entrene, es decir, debemos de procesar los datos que le damos a nuestro algoritmo para que este pueda exprimirlos al máximo. Vamos a ver los tres procesos que llevan a los errores más comunes en este tipo de proyectos (obviamente hay muchas más comprobaciones y pruebas que no vamos a tratar por ahora). 

Para entender bien lo que vamos a hacer ahora debemos de tener claro el **objetivo** que queremos que aprenda nuestro modelo. Nosotros no queremos saber cómo de buena va a ser una empresa de forma aisalda sino más bien, de todo el conglomerado de empresas que tenemos a nuestra disposición, cuál de todas ellas lo va a hacer mejor en comparación al resto en el próximo mes. Lo mismo pasa con los factores, nosotros no queremos predecir el momentum de una empresa sino saber cuál tendrá mayor momentum en un momento dado. 

---

#### **Normalización cross-sectional**

Recordando la misma idea del título del proyecto, a nosotros lo que nos interesa no es la evolución temporal de un activo sino las relaciones entre diferentes activos en un instante dado (por eso lo de cross-sectional). Cuando hablamos de normalización simplemente es coger un mismo factor calculado para una fecha concreta de todas las empresas y normalizamos. 

> Poniendo de ejemplo el momentum, que se calcula mensualmente, cada mes obtendrías un nuevo dato de este factor para cada empresa y volerías a normalizar. Es decir, cada fecha se trata por separado. 

**¿Por qué no normalizar utlizando toda la historia?** 

Porque un factor puede considerarse alto o bajo dependiendo del contexto donde nos encontremos. En el caso de la volatilidad, no es lo mismo que un activo tenga mucha volatilidad en un periodo tranquilo donde la "volatilidad base" está muy baja a que presente estos mismos números en un contexto de gran incertidumbre y volatilidad estructural por las nubes. 

> Los factores se analizan comparando empresas actuales entre sí, no a lo largo del tiempo. 

¿Cómo normalizamos "crossectionalmente" un factor?

Cómo la magnitud del factor de una empresa no es lo que importa sino más bien la relación entre magnitudes de varias empresas, vamos a comvetir los números absolutos en rankings o percentiles. Existen varias maneras de hacer esto: 

- **z-score**: se trata de obtener el promedio de un factor de todas las empresas y ver para cada una de ellas cómo de alejadas, en desviaciones estándar, están de la media. Ahora ya no tenemos magnitudes absolutas sino cómo de por encima o por debajo está una empresa respecto de la media en este factor. 

- **rank percetil**: se trata de colocar en orden ascendente, según la magnitud del factor, todas las empresas. Luego podemos normalizar a 100 y tendríamos la posición de cada empresa en cada factor como un percentil. 

> Normalmente se suele utilizar más el rank percentil ya que muchos factores financieros suelen seguir distribuciones no gaussianas, con colas pesadas, por lo que tanto la media como la desviación estándar se verían muy afectadas por estos outliers, afetando así al z-score (que se calcula directamente a partir de estas dos magnitudes). En cambio en el rank percetil, por muchos outliers que haya, la manera de ordenarlos es por posición y el ranking no depende de otras magnitudes sensibles a heavy tails.

---

#### **Winsorización**

La winsorización es una manera de tratar los valores outliers, sucesos que son bastante comunes en los mercados financieros, para que no estropeen toda la normalización. Lo que hacemos es fijar una cota de percentil, por ejemplo del percentil 1 al percentil 99, y rebajamos el dato extremo hasta dicha cota. Por lo tanto todos los outliers por encima del percentil 99 y por debajo del percentil 1 pasarían a tener el valor del percentil 99 o del percentil 1. 

> Lo que hacemos con esto es evitar que un único dato extremo domine la distribución modificando deliberadamente los datos y así conseguir mayor robustez. 

> La cota de percentiles puede variar pero 1-99 y 5-95 suelen ser los más usados. 

**¿Por qué no eliminar el outlier directamente?**

Podría ser que el outlier es simplemente un dato mal tomado y convendría quitarlo pero en la mayoría de los casos sí que responde a que realmente esa empresa lo hizo extraordinariamente bien (o extraordinariamente mal) en ese mes. La clave está en que la winsorización no intenta decir que el dato extremo sea falso sino evitar que un único dato tenga un peso desproporcionado en el aprendizaje del modelo. 

**¿Cuándo aplicar la winsorización?**

La winsorización se aplica siempre antes de normalizar. En el caso de utilizar el z-score, siempre va a ser necesario winsorizar para no desestabilizar las métricas de la distribucición. Para el caso del rank percentile tienes dos opciones, cada uno con sus pros y contras: 

1. **Winsorizas**: evitas trabajar con outliers tan bestias (que a la larga pueden dar problemas) pero todas las empresas con valores extremos empatan en el percentil más alto y se complica el orden. En esta opción estaríamos priorizando una mayor robustez frente a posibles errores o valores extremos pese a que exista una pérdida de información. 

2. **No winsorizas**: no te quitas los ouliers pero facilitas el orden de las empresas con valores extremos. Podríamos defender esta opción también argumentando que el rank percentil ya de por sí reduce muchísimo el impacto de outliers y estaríamos conservando más información que en el primer caso. 

> En este tipo de dilemas donde la solución teórica no es clara lo mejor que podemos hacer entrenar exactamente el mismo modelo con ambas opciones y ver en el backtest cuál de las dos arroja mejores resultados. 

> Normalmente no es tan importante que una empresa esté en el ranking 1 o en el ranking 3 sino más bien saber que pertenece al grupo de, por ejemplo, momentum extremadamente alto. 

---

#### **Lag correcto**

Sería otra forma de evitar utilizar información que en ese momento todavía no existia (otra manera de look-ahead bias). Se basa en el hecho de que la publicación trimestral de resultados de las empresas (los datos fundamentales) no se sabe el día siguiente al cierre del trimestre sino semanas después. Por mucho que el trimestre haya concluido, la empresa ecesita tiempo para cerrar la contabilidad, revisar las cuentas, auditarlas, preparar el informe y presentarlo oficialmente por lo que sería un error que desde el día uno comenzásemos a utilizar el dato nuevo ya que nadie conoce oficialmente esa información.     

> El lag retrasa artificialmente la disponibilidad del dato. 

Este problema es relevante solamente si utilizamos factores que involucren informaciones como beneficios, deuda, márgenes, etc. pero los factores que se puedan constuir directamente a partir de los precios o volúmenes sí que son instantáneos (al día siguiente se sabe el precio con el que se cerró la cotización de ayer). 

> Existen empresas especializadas como Bloomberg o FactSet que no solo almacenan el valor de una variable sino cuándo se generó, cuándo se publicó, o cuándo estuvo disponible para el mercado.

## Fase 3 — Definición del target y construcción del dataset de entrenamiento (3-4 días)

- **Target**: forward return a 1 mes, idealmente convertido a rank percentil cross-sectional en cada fecha (esto convierte el problema en uno de *ranking relativo* más que de predicción exacta del retorno, que es mucho más difícil y menos útil — en la práctica lo que importa para construir cartera es el orden relativo, no el valor exacto).
- **Formato final**: una tabla con (fecha, ticker, factor_1, ..., factor_k, target) lista para alimentar al modelo, con un cuidado especial en que el target de la fila correspondiente a la fecha *t* use información solo hasta *t*, y el retorno objetivo sea estrictamente *t+1 mes*.
- **Missing data**: decidid una política explícita (¿eliminar filas con missing, o imputar con la mediana cross-sectional de ese factor en esa fecha?) y documentadla — es otro punto donde los entrevistadores preguntan "¿y qué hicisteis con los missing?" y "no nos dimos cuenta de que había" es la peor respuesta posible.

---
---

#### **Fornato final**

Es muy importante tener en cuenta lo que le vas a pasar al modelo para que funcione, es decir, el algoritmo no entiende de factores sino de números. El formato estándar es una tabla donde cada fila contiene toda la información de una empresa en particular (fecha, ticker, momentum, volatilidad, tamaño, liquidez, target, etc.) obviamente información ya procesada y normalizada como hemos explicado anteriormente. Hemos pasado de construir variables a constuir **ejemplos de entrenamiento**. 

> El **target** es la variable que el modelo intenta predecir, en nuestro caso, el rendimiento de una empresa el mes siguiente. 

Lo que hará el modelo es detectar relaciones entre los diferentes factores de una empresa y su target para predecir a partir de miles de ejemplos las magnitudes de los factores que más se relacionan con maximizar el target. 

> Es muy importante introducir un lag a futuro sobre el target, es decir, con los factores de hoy queremos predecir el target de mañana, no el de hoy. 

#### **Missing data**

Un **missing** es un dato que falta. Las razones pueden ser por una mala descarga de datos, porque la empresa ha salido recientemente a cotizar y no existe información todavía o porque no se han presentado resultados. En estos casos existes dos soluciones: 

1. **Eliminar la fila**: si una empresa no presenta un factor, ese mes directamente operas como si esa empresa no existiese. El problema aquí es que te estás comiendo muchísima información. 

2. **Imputar**: se trata de rellenar el dato que falta utilizando alguna regla previamente definida. La opción más comun es rellenar el dato con la mediana cross-sectional del factor (utilizamos solamente esa fecha, no el histórico) obtenida del resto de empresas.  

> Se utiliza la mediana en vez de la media porque es mucho más robusta frente a outliers. En cambio la media puede variar demasiado. 

> Cualquiera de las dos opciones son válidas. La decisión de tomará dependiendo del caso: si la empresa todavía no contizaba podría ser buena idea suprimirla en esa fecha pero si el dato está mal cogido sería mejor idea imputarlo.

## Fase 4 — Metodología de validación temporal (clave, no os la saltéis) (3-4 días de diseño + implementación)

Este es probablemente el punto más importante de todo el proyecto desde el punto de vista de "esto demuestra que sabéis lo que hacéis".

- **Nunca k-fold aleatorio**: mezclar fechas aleatoriamente entre train y test permite que el modelo "vea" información del futuro indirectamente (por ejemplo, dos meses consecutivos del mismo activo están correlacionados, así que si uno cae en train y el contiguo en test, hay fuga).
- **Walk-forward validation**: entrenad con una ventana de datos (por ejemplo 5 años), validad/testead en el periodo inmediatamente siguiente (por ejemplo el siguiente trimestre o año), y luego desplazad la ventana hacia delante, repitiendo el proceso. Esto simula cómo se usaría el modelo en producción: solo con información pasada.
- **Embargo/purga**: dejad un pequeño hueco temporal entre el final del train y el inicio del test (por ejemplo el horizonte del target, 1 mes) para evitar fuga por el solapamiento de las ventanas de cálculo de factores y targets. Esto es una idea que viene de Marcos López de Prado (*Advances in Financial Machine Learning*) — si lo mencionáis en una entrevista, da una señal muy fuerte de que habéis leído literatura seria del campo, no solo blogs.
- **Holdout final**: el último 1-2 años de los que hablábamos en la fase 0 no se toca durante todo el desarrollo del modelo. Solo se usa una vez, al final, para el backtest definitivo. Si os encontráis "espiando" ese periodo durante el desarrollo (aunque sea para tomar decisiones de diseño), perdéis la validez de esa evaluación final — sed disciplinados con esto.

**División de trabajo**: este es un buen punto para hacerlo juntos los dos, porque es conceptualmente denso y conviene que ambos lo entendáis a fondo — es justo el tipo de cosa que os van a preguntar en una entrevista técnica y "lo hizo mi compañero" no es una respuesta aceptable aquí.

**¿Cómo demuestro que mi modelo funciona?**

Algo incluso más importante que constuir el modelo va a ser asegurarnos de que está funcionando adecuadamente.  El punto clave aquí es que en ningún momento el modelo puede aprender utilizando información, que, en la vida real, todavía no habría existido. 

**¿Qué es validar un modelo?**

La manera lógica sería, como hemos visto, reservar una parte de los datos para el entrenamiento y otra totalmente independiente para comprobar si funciona. El problema de esta técnica es que la partición que haces es totalmente arbitraria y podría pasar que la parte de prueba sea más "fácil" que la parte de entrenamiento por lo que al comprobar si el modelo funciona verías que lo hace demasiado bien. 

**k-fold**: se trata de hacer la división prueba-entrenamiento de muchas maneras diferentes asegurándote así de que todos los datos pasan alguna vez por el entrenamiento y por la prueba. Así estaríamos reduciendo mucho la dependencia de una única partición, obteniendo un promedio de estimación mucho más estable.

#### **Nunca k-fold aleatorio**

En nuestro caso no podemos utilizar el k-fold bajo ningún concepto porque estamos tratando con series temporales, es decir, no toda la información se trata de la misma manera ya que existe una flecha temporal hacia el futuro que te dicta una dependencia de valores futuros respecto de los valores pasados. Si utilizamos el k-fold aleatorio estaríamos metiendo en el entrenamiento datos del futuro por lo que implícitamente el modelo aprende patrones que solo existen en el futuro. 

Por tanto, en nuestro caso, debemos de obedecer la siguiente regla: nunca entrenar con observaicones posteriores al periodo que estamos intentando predecir. 

> En el Machine Learning cláscio normalmente se asume que las filas del dataset son independientes pero en una serie temporal no es así. Pueden haber ddependencias temporales, autocorrelaciones, tendencias, etc. 

#### **Walk-forward validation**

Como no podemos utilizar el k-fold, para no depender de una única división arbitraria utilizamos un walk-farward validation donde siempre utilizaremos una ventana temporal de entrenamiento anterior a la ventana temporal de prueba. Esta validación tiene dos variantes: 

1. **Rolling window**: cogemos siempre la misma ventana temporal para entrenar el modelo y la vamos "deslizando" a lo largo de los años.
2. **Expanding window**: cada vez entrenas el modelo con una ventana temporal mayor. 

> El walk-forward es realmente cómo trabajaríamos día a día con el modelo ya que mes tras mes irías actualizando la ventana temporal de entrenamiento para incluir la nueva información que te va llegando. 

#### **Purga**

La **purga** consiste en eliminar del Training Set los datos cuyo período de observación se solapa con el inicio del Test Set. Esto se hace porque la validación del dato en el entrenamiento, si consideramos nuestro caso, no se ve instantáneamente sino que tiene un lag de un mes por lo que si el entrenamiento del modelo termina en diciembre y el test comienza en enero, todos los datos de diciembre se estarían validando con los rendimientos de enero (el dato del 20 de diciembre se valida el del 20 de enero). 

Esto no es una fuga tan grave como entrenar con datos del futuro pero si no la tenemos en cuenta habría información compartida entre los dos conjuntos y eso puede hacer que el rendimiento parezca ligeramente mejor de lo que realmente sería. 

#### **Embargo**

El **embargo** consiste en eliminar del Test Set (o del bloque posterior) los datos que sufren de autocorrelación con el final del entrenamiento. Incluso después de purgar, los mercados financieros tienen "memoria" a través de un fenómeno llamado autocorrelación serial (los rendimientos de hoy se parecen a los de ayer, la volatilidad persiste, etc.). Si el entrenamiento acaba justo antes del Test, los patrones más recientes que vio el modelo a finales de noviembre siguen muy "frescos" y presentes en los primeros días de enero. El modelo podría tener una ventaja artificial en las primeras predicciones de Test simplemente porque el régimen de mercado no ha cambiado por lo que se introduce un "periodo de descanso" o "cuarentena" donde no se realizan predicciones para dejar que la memoria del mercado se disipe.

> Si queremos empezar a predecir en enero, la purga sería todo el mes de diciembre (queda en el lado del entrenamiento) y el embargo los primeros quince días de enero (queda en el lado del test)

> Si metemos las tres fases de **train, validation y test** introduciríamos una purga y un embargo entre train y validation y otras dos entre validation y test. 

**¿Si aislamos cada conjunto utilizando la purga y el embargo, por qué seguir mirando solo hacia adelante?**

Los grandes problemas del **walk-forward validation** vienen por la **inestabilidad de los resultados** ya que solo pruebas cada periodo histórico una vez y por el **desperdicio brutal de los datos** ya que te obliga a usar rutas temporales fijas hacia adelante. Si analizamos la razón por la cual descartamos el k-fold nos podemos dar cuenta que precisamente la purga y el embargo nos elimina esa restricción temporal. 

**Combinatorial Purged Cross-Validation (Lopez de Prado)**

Lopez de Prado se dio cuenta de que si aplicas Purga y Embargo de forma matemática por ambos lados de un bloque, ya no necesitas obligar a que el entrenamiento esté estrictamente antes en el tiempo que el test. El CPCV precisamente explota esa idea diviendo la serie temporal en N bloques (por ejemplo 6 bloques de 2 años) y dejando que el algoritmo coja todas las combinaciones posibles entre train y test sin fijarse en qué bloques van antes o después. Lo que haremos será: 

- CPCV para el ajuste de hiperparámetros (Tuning): cuando quieras encontrar la mejor profundidad de árboles o la tasa de aprendizaje para tu XGBoost esta técnica te permitirá testear miles de combinaciones aprovechando al máximo todos los años de datos que tengas sin sobreajustar.

- Usa Walk-Forward (con Purga + Embargo) para la simulación final (Backtest): una vez que ya has elegido el modelo con sus hiperparámetros definitivos mediante CPCV, haz el "examen final" simulando un Walk-Forward estricto hacia el futuro. Esto es lo que le enseñará al mundo que tu modelo es capaz de operar en la vida real paso a paso sin mirar atrás.

## Fase 5 — Modelado (1.5-2 semanas)

**Empezad simple, luego añadid complejidad:**

1. **Baseline**: regresión lineal o Elastic Net sobre los factores normalizados. Esto os da un punto de referencia interpretable (los coeficientes os dicen directamente qué factores pesan más) y es rápido de entrenar en cada ventana del walk-forward.
2. **Modelo no lineal**: LightGBM o XGBoost (LightGBM suele ser más rápido para forecasting tabular tipo panel). Estos modelos capturan interacciones entre factores que la regresión lineal no puede.
3. **Comparación honesta**: si el modelo no lineal no mejora sustancialmente al baseline lineal (es un resultado muy común en finanzas, donde la señal es débil y ruidosa), decidlo y explicad por qué puede estar pasando, en lugar de forzar una conclusión de que "el ML funciona mejor" sin evidencia.

**Métricas de evaluación del modelo en sí (antes de construir cartera):**

- **Information Coefficient (IC)**: correlación de Spearman entre las predicciones del modelo y los retornos reales observados, calculada en cada fecha de rebalanceo y promediada. Es la métrica estándar en la industria para evaluar el poder predictivo de un modelo de factores, mucho más relevante que el R² o el RMSE en este contexto.
- **Decay del IC**: ver cómo se degrada el poder predictivo si alargáis el horizonte (1 mes, 3 meses, 6 meses) — os da intuición de cuánto "dura" la señal.
- **Feature importance / SHAP**: usad SHAP values sobre el modelo de árboles para poder explicar qué factores está usando el modelo y cómo, no solo que "funciona". Esto da pie a una sección de interpretabilidad en el README que casi ningún proyecto de estudiante incluye bien.

**¿Qué algoritmo utilizamos para aprender la relación entre los factores y el target?**

Nosotros no queremos demostrar que un modelo complejo de ML funciona bien sino que funciona claramente mejor que un modelo sencillo. Es decir, antes de meternos en un XGBoost o similar debemos establecer un **baseline** y luego ya ver qué tal lo hacen los demás modelos en comparación. 

**Baseline** 

Como baseline utilizaremos modelos extremadamente interpretables y sencillos, estos van a ser la Regresión Lineal y el Elastic Net: 

- **Regresión Lineal de Mínimos Cuadrados (OLS)**: asume que la relación entre tus variables (Momentum, Tamaño, Volatilidad) y el rendimiento futuro de la acción es completamente recta y directa. Si una regresión lineal no es capaz de predecir nada, te da una métrica clara de lo ruidosos que son tus datos y al compararla con modelos más complejos sabrás exactamente cuánta ventaja estás ganando gracias a capturar patrones no lineales.

> El problema de la Regresión Lineal es que sufre muchísimo en finanzas porque las variables van a tener multicolinealidad y el modelo se volverá inestable.

- **Elastic Net**: sería como el baseline inteligente (como una regresión con esteroides) que introduce un mecanismo de regularización que combina Lasso (L1), que pone a cero las variables que no importan para bajar el número de variables, y Ridge (L2), que redistribuye la importancia entre variables que miden casi lo mismo para reducir los números gigantescos. 

> ElasticNet es la mejor aproximación lineal linealmente estable por lo que si tras ajustar ElasticNet, un XGBoost o un Random Forest consiguen batirlo por un margen claro, habrás demostrado matemáticamente que en tu problema existen interacciones complejas y no lineales entre las variables que un modelo tradicional jamás habría podido detectar.

En finanzas es bastante común encontrar estudios donde el modelo sencillo funciona igual de bien porque la información contenida en esos factores quizá no justifica un modelo tan complejo. 

---

**¿Por qué las finanzas son un problema tan difícil para el ML?**

Merece la pena pararnos a entender por qué la IA no ha "resuelto" los mercados financieros, a pesar de haber revolucionado casi todos los demás campos. El punto principal es que las finanzas son uno de los peores escenarios posibles para aplicar Machine Learning, no porque los modelos sean malos sino porque el problema es extremadamente difícil, es decir que, el cuello de botella casi nunca es el algoritmo sino más bien la calidad y la cantidad de información predictiva disponible.
Por eso es muy fácil que ocurra que la información contenida en los factores sea tan débil que muchas veces un modelo muy sofisticado no tiene mucho más que aprender que uno sencillo. Los principales problemas son: 

- **Señal extremadamente débil**: los factores que manejamos contienen muy poca información (con ellos se explica solamente un pequeño porcentaje del rendimiento) porque casi todo es puro ruido. 

- **El mercado cambia constantemente (non-stationarity)**: los patrones y relaciones aprendidos en un periodo de tiempo pueden dejar de cumplirse o desaparecer. 

- **Los propios modelos cambian el mercado**: si alguien publica una nueva estrategia, el propio éxito del modelo contribuye a destruir el propio patrón que había descubierto. 

- **Existen muchos factores ocultos**: una subida inesperada de tipos, una pandemia o una guerra introducen factores inesperados que hacen que los modelos pierdan capacidad predictiva. 

- **El mercado ya es muy eficiente**: al haberse profesionalizado las relaciones evidentes o "más fáciles" ya han sido explotadas. 

- **El sobreajuste es extremadamente fácil**: al haber tantos factores y tantos hiperparámetros que ajustar, es muy fácil caer en una combinación que por pura causalidad sea increiblemente rentable. 

La **única ventaja** es que aunque la señal sea muy pequeña, si consigues capturar un poquito puede ser muy rentable. En la bolsa no se busca acertar el 100% de las veces (como puede ocurrir en el reconocimiento facial o en la detección de tumores) sino mejorar ligeramente el azar. 

---

**¿Cómo sé que el modelo funciona?**

Antes de constuir la cartera debemos de asegurarnos de que el modelo ha alcanzado con éxito el objetivo propuesto. No nos olvidemos que este no es predecir el rendimiento de un activo sino ordenar los activos en función de los rendimientos previstos. Por eso no tendría sentido mirar, por ejemplo, el RMSE (Root Mean Squared Error) porque nosotros queremos que el modelo sepa ordenar acciones, nada más. 

- **IC (Information Coefficient)**: valora el grado en que el modelo ha acertado a la hora de rankear los diferentes activos en función de los retornos. No habla ni de dinero, ni de pesos, ni de costes, ni de riesgo, simplemnete del ranking. El IC puede variar entre 1 (ranking idéntico a la realidad) y -1 (ranking totalmente inverso) pero en proyectos reales nos solemos mover alrededor de 0.02-0.06 precisamente por el ruido gigantesco que manejamos. 

> Realmente existen dos IC: el Rank IC (Spearman) que es al que se utiliza en este caso y al que se refiere todo el mundo cuando habla de "Information Coefficient" y el Normal IC (Pearson) que mide la relación lineal directa entre los valores numéricos exactos de la predicción y los rendimientos.

> El modelo termina justamente aquí, toda la parte de contrucción de la cartera (los pesos que les otorgas a las empresas, cuántas tienes en cuenta, etc.) entra en otra parte del proyecto. Es decir, una cosa es detectar empresas prometedoras y otra muy distinta es convertir esas predicciones en una buena estrategia de inversión. 


- **Decay del IC**: simplemente nos dice cómo de rápido se va a ir desgastando nuestra predicción del ranking si alargamos el horizonte temporal. Normalmente se calula a uno, tres y seis meses y nos da una medida directa de cuánto tarda nuestra predicción en perder utilidad (la persistencia en el tiempo). 

- **Information Ratio del IC ($IR_{IC}$)**: 

> El decay nos sirve para saber si deberíamos de rebalancear muy a menudo o no. Sería como una métrica para hacernos una idea de la frecuencia de rebalanceo que debemos de manejar. 

> También nos sirve como una comprobación de coherencia, es decir, si el modelo está entrenado para predecir a un mes vista, el decay al alargar el horizonte temporal debe de ser evidente. 

- **Feature importance**: nos dice cuáles son las variables qué más utiliza el modelo para tomar decisiones en general. 

- **SHAP**: nos dice cómo ha contribuido cada factor a la predicción concreta de una empresa, es decir, reparte la "responsabilidad" de la predicción entre las distintas variables. 

> En la práctica se suele utilizar el Featura Importance para una visión global y el SHAP para un análisis más profundo. 

> Estas métricas son super importantes en el READ ME porque demuestran si nos hemos fiado ciegamente de los resultados aportados por el modelos o si realemente entendemos de dónde salen esas decisiones. 

---

**¿Cómo de bien debe el modelo ordenar las acciones?**

Una vez sabemos las métricas que se utilizan debemos de establecer los criterios que nuestro modelo debe superar para considerar que "ha ordenado bien". Estos exámenes se harán obviamente en la parte del Test / Out-of-Sample:

- **Rank IC medio > 0.03**: el promedio del rank IC calculado a lo largo de todos los meses debe estar por encima de ese mínimo exigible. 

- **Information Ratio del IC ($IR_{IC}$) > 0.5 (anualizado)**: el ratio entre el Mean Rank IC y la desviación estándar de los IC mensuales debe garantizar que las predicciones del modelo son consistentes y estables mes tras mes. 

También debemos de tener en cuenta la **calidad del modelo**: los retornos deberían aumentar casi de forma monótona conforme subimos en el ranking (curva descendente casi perfecta) ya que siginificaría que el modelo es capaz de ordenar bien todo el universo. Si el modelo no ordena bien por dentro estaríamos optimizando sobre un ranking "roto".

## Fase 6 — De predicciones a cartera (1-1.5 semanas)

Aquí entra directamente lo que ya sabes de Markowitz, pero aplicado de forma más realista:

- **Construcción simple (punto de partida)**: en cada fecha de rebalanceo, ordenad los activos por la predicción del modelo y formad carteras por deciles (long el decil superior, short el decil inferior si haces long-short; o simplemente sobreponderar el decil superior si preferís long-only, que es más realista si alguno de los dos no tiene acceso/experiencia con posiciones cortas).
- **Construcción con optimización (más interesante)**: en lugar de pesos iguales dentro de cada decil, usad las predicciones del modelo como inputs de retornos esperados en una optimización media-varianza (con la matriz de covarianza estimada de forma robusta, por ejemplo con shrinkage tipo Ledoit-Wolf en lugar de la covarianza muestral pura, que es ruidosa con pocos datos). Esto conecta directamente con la teoría de carteras de los 50 que estás estudiando, pero usando como input no los retornos históricos esperados (que es la debilidad clásica de Markowitz, "garbage in garbage out") sino las predicciones del modelo de ML — es una forma elegante y defendible de unir las dos partes del proyecto.
- **Restricciones realistas**: límite de peso máximo por activo (para no concentrar toda la cartera en 2-3 nombres), límite de turnover por rebalanceo (para controlar costes), y si hacéis long-short, neutralidad de exposición neta o sectorial si os da tiempo.

---
---

Una vez tenemos el modelo entrenado debemos de tranformar ese ranking en una cartera de acciones. Para ello vamos a tener que responder a varias cuestiones que involucran la proporción del ranking que incluiremos en nuestra cartera y los pesos que les otorgaremos a las empresas en cada caso. 

**¿Cuál es nuestro benchmark?**

Antes de empezar a optimizar la cartera debemos de saber con certeza que nuestro modelo es capaz de ordenar correctamente las acciones porque si no es así ninguna optimización posterior va a ser capaz de rescatar una señal que no contiene información. El benchmark o punto de partida que introducimos va a ser el **long-only equal weight sobre el decil superior**, es decir, cogemos el mejor 10% del ranking y le otorgamos a cada acción el mismo peso. 

### **Porción del ranking en nuestra cartera**

1. **Corte fijo por cuantil (Long-only)**: tomas la lista ordenada y aplicas un tijeretazo en un porcentaje fijo del ranking en cada periodo de rebalanceo. 

    - **El decil superior**: maximizas el Alpha esperado pero tienes mayor volatilidad y rotación alta si la posición de las acciones oscila mucho cerca del límite del corte.

    - **El quintil superior**: mayor diversificación y estabilidad pero diluyes ligeramente la potencia de la señal. 

2. **Estrategia Long/Short o Market-Neutral (Doble Corte)**: si en lugar de querer solo comprar acciones buscas una estrategia que gane dinero independientemente de si el mercado sube o baja, lo que haces es beneficiarte del diferencial de rendimiento entre lo que tu modelo considera "excelente" y lo que considera "basura". 

    - **Long/Short por Deciles**: compras (Long) las acciones del decil 10 y vendes en corto (Short) las acciones del decil 1. 

3. **Umbral por Puntuación Absoluta (Score Threshold)**: en lugar de cortar de forma rígida por un porcentaje fijo, estableces una regla basada en el valor numérico de la predicción de tu modelo. Es decir, cada mes solo entrarían en la cartera las acciones cuya rentabilidad esperada predicha por el modelo supere un listón concreto. Lo que consigues es adaptar el tamaño de tu cartera al entorno del mercado. 

4. **Estructura con Zonas de Amortiguación (Buffer Zones)**: para evitar que las empresas en la parte baja del decil superior estén entrando y saliendo constantemente (lo que se trasladaría a costes de transacción absurdos) se introduce una lógica de bandas de tolerancia. Podría ser por ejemplo que para que una acción entre en cartera debe superar el 10% mejor pero para salir de cartera debe caer por debajo del 15% mejor. 

**¿Long-Only o Long-Short?**

El problema de la estrategia Long-Only es que siempre vamos a tener dos efectos mezclados ya que el resultado es la combinación del mercado y de la calidad del modelo (si el mercado es alcista y ganamos dinero no necesariamente significa que el modelo sea bueno). Precisamente para aislar la calidad del modelo de la exposición al mercado se utiliza la estrategia **Long-Short** ya que cuando el mercado sube las posiciones largas ganan y las cortas pierden y cuando el mercado baja las posiciones largas pierden y las cortas ganan, como si gran parte del movimiento del mercado se cancelase. 

Idealmente la rentabilidad de esta estrategia debería ser integramente gracias a la calidad del modelo ya que el efecto del mercado ha sido eliminado, a esta de aislarse por completo de los movimientos del mercado para ganar dinero únicamente por la habilidad de selección de activos se le llama **Market Neutral**. Por eso casi toda la investigación académica usa long-short, ya que una cartera de este tipo es capaz de reducir mucho la beta (parte explicada por el movimiento general del mercado) permitiendo observar con más claridad el alpha (exceso de rentabilidad atribuible a la calidad de la estrategia o del modelo) generado por la señal. 

> En la práctica las posiciones cortas tienen inconvenientes como los costes por tomar prestadas las acciones o las pérdidas potenciales. 

> Normalmente se utiliza la Long-Short para evaluar la calidad pura del modelo (si la señal tiene realmente poder predicitivo) y la Long-Only para evaluar la utilidad práctica de la estrategia para un inversor. 




### **Distribución de pesos en los activos**

1. **Asignación Heurística**: son métodos simples y robustos que no requieren matrices complejas de covarianza ni optimizadores matemáticos. Su principal ventaja es que no sobreajustan y son muy difíciles de batir en la práctica.

    - **Equal-Weighted (Equiponderada)**: le das exactamente el mismo porcentaje a cada acción. Es la opción más sencilla de implementar.
    
    - **Signal-Weighted (Ponderación por Fuerza de la Señal)**: el peso es proporcional a la predicción del modelo, es decir, las accione con mejores rendimientos ponderan en mayor medida. Esto exprime al máximo la capacidad predictiva del modelo pero si la predicción tiene ruido o errores puntuales, el error se magnifica en la cartera.
    
    
2. **Asignación Basada en Riesgo (sin estimar retornos)**: en lugar de confiar a ciegas en las predicciones del modelo para fijar los pesos, estas técnicas usan el modelo solo para seleccionar las acciones, y usan el riesgo (la volatilidad o correlación) para calcular los pesos.

    - **Volatility-Weighted / Inverse Variance (Volatilidad Inversa)**: el peso de cada acción es inversamente proporcional a su volatilidad reciente. Reduce drásticamente la volatilidad total de la cartera y suaviza los drawdowns. 
    
    - **Risk Parity (Paridad de Riesgo dentro del Decil)**: ajusta los pesos considerando no solo la volatilidad individual de cada acción, sino también cómo están correlacionadas entre sí. El objetivo matemático es que cada acción aporte exactamente la misma cantidad de riesgo al total de la cartera evitando así que acciones que se mueven igual dominen todo el riesgo.
    
3. **Optimización Matemática (Markowitz y derivados)**: aquí le entregas al algoritmo el retorno esperado (predicho por el modelo) y la matriz de covarianza de las acciones.

    - **Max Sharpe / Mean-Variance Optimization (Markowitz)**: busca la combinación exacta de pesos que maximiza el Ratio de Sharpe estimado. Siempre se debe usar con restricciones.
    
    - **Black-Litterman**: parte de una cartera de referencia neutra (por ejemplo, por capitalización de mercado) y ajusta suavemente los pesos hacia las acciones que el modelo prefiere, considerando la incertidumbre/confianza que tienes en la señal. Es la técnica estándar en fondos cuantitativos institucionales porque evita los pesos extremos y arbitrarios de Markowitz.

**¿Cómo conviertes el ranking que te da el modelo en retornos esperados?**

Si queremos obtener los pesos a través de la optimización matemática necesitamos darle de comer al algoritmo los retornos esperados predichos por el modelo, es decir, debemos convertir el score en retornos esperados mediante **calibración**, que básicamente se basa en relacionar cada score con el retorno medio, observado después. También podríamos hacer que el propio modelo prediga los retornos cambiando el target al Forwar Return en vez de el Ranking pero sería un problema mucho más inestable. 

Sabiendo que el retorno esperado se compone de alpha más beta, podemos directamente pasarle al optimizador el alfa como vector de rentabilidades y la matriz de covarianzas. De esta manera le estaríamos diciendo al optimizador que maximice no el rendimiento general sino el extra que nos da nuestra ventaja predictiva, consiguiendo carteras que buscan inmunidad al mercado. Esto se traslada a que, en lugar de maximizar el ratio Sharpe (que incluye el riesgo del mercado) estaríamos maximizando el **Information Ratio** (sacamos el máximo partido a la señal predictiva por cada unidad de riesgo propio que asumimos).

> Con esto ya de paso estaríamos resoviendo otro de los problemas que tiene la teoría clásica de Markovitz, la cual utiliza como retorno el histórico medio con los problemas que acarrea. 

**¿Qué pasa si todas las empresas del decil superior se mueven igual?**

A pesar de creer que estamos diversificados porque tenemos 15 empresas en nuestra cartera, puede ocurrir que estas pertenzcan a un mismo sector por lo que al igual que ahora todas suben en algún momento corregirán juntas (al igual que está pasando ahora con el sector tecnológico). La idea de la diversificación no es tener muchos activos sino que estén poco correlacionados. El problema del ranking es precisamente que solo se responde a qué empresas van a comportarse mejor pero no cómo de parecidas son entre sí. 

El algoritmo de optimización tiene en cuenta tanto el rendimiento como la matriz de covarianzas y podría pasar que el mejor activo no forme parte de la cartera ya que el riesgo que aporte en esta es muy grande debido a su correlación con el resto de activos. 

Esta es una de las razones por las que la construcción del modelo y la contrucción de la cartera pertencen a modulos totalmente separados, en cada caso queremos optimizar algo diferente. Tendríamos dos fuentes de valor completamente distintas: la calidad de la señal (alpha) con la que encontramos las mejores oportunidades y la calidad de la contrucción de cartera con la que aprovechamos esas oportunidades de la forma más eficiente. 


**¿Qué matriz de covarianzas le paso al optimizador?**

Ya tenemos claro de donde salen los retornos (o más bien el alpha) que le pasamos al optimizador pero todavía no hemos hablado de la matriz de covarianzas. Una primera idea sería introducir la **matriz de covarianzas muestral** que calcula las varianzas y correlaciones basándose únicamente en los datos históricos. El problema está en que para que esta matriz sea matemáticamente estable y fiable, necesitarías muchísimos más datos temporales (filas) que activos en la cartera (columnas), pero en la realidad ocurre justo lo contrario, tenemos pocas observaciones en el tiempo y muchas acciones. Esto provoca dos desastres: 

- **Inversión de Matrices Instable**: los optimizadores matemáticos necesitan invertir la matriz de covarianza para calcular los pesos por lo que cualquier ruido se multiplica. 

- **Error de Estimación Extremo**: la covarianza muestral sobreestima las correlaciones más altas y subestima las más bajas solo por puro azar (al buscar la máxima diversificación, el obtimizador se enamora de las de las correlaciones falsamente bajas y mete un peso absurdo). 


**Estimación de Ledoit-Wolf (concepto de Shrinkage)**: en lugar de confiar a ciegas en los datos históricos (covarianza muestral) o usar un modelo teórico demasiado simple, Ledoit y Wolf (2003/2004) introdujeron la técnica del Shrinkage (encogimiento o suavizado) creando una combinación lineal ponderada entre dos matrices:

$$\Sigma_{\text{Ledoit-Wolf}} = (1 - \delta) \cdot \mathbf{S} + \delta \cdot \mathbf{F}$$

- $\mathbf{S}$ (Covarianza Muestral): Tus datos reales. Tienen mucha información útil pero también mucho ruido.

- $\mathbf{F}$ (Matriz Objetivo o Target): Una matriz teórica muy simple y estructurada (por ejemplo, asumiendo que todas las acciones tienen la misma correlación entre sí). Tiene cero ruido, pero es demasiado simplista.

- $\delta$ (Intensidad de Encogimiento / Shrinkage Intensity): Un número entre $0$ y $1$.

Ledoit-Wolf calcula matemáticamente de forma automática el valor óptimo de $\delta$ para minimizar el error cuadrático medio esperado de la matriz. Si tienes pocos datos o mucho ruido, el algoritmo sube el $\delta$ y se apoya más en la matriz estructurada $\mathbf{F}$ para evitar que el optimizador se vuelva loco y si tienes muchos datos limpios, el algoritmo baja el $\delta$ y le da más peso a los datos reales de $\mathbf{S}$.

**¿Qué restricciones le ponemos al optimizador?**

A parte de las básica (la suma de los pesos deben dar 1) necesitamos introducir una serie de restricciones para que el algoritmo no haga locuras: 

- **Peso máximo por activo (Maximum Position Size)**: limitamos al 3%-5% la porción de nuestro portfolio para una sola empresa. Así evitamos que un crash inesperado nos reviente la cartera.

- **Peso mínimo por activo**: limitamos al 0.5% para no tener decenas de mini porciones que prácticamente no aportan rentabilidad pero que generan costes y seguimiento administrativo. 

- **Long-Only**: si queremos este tipo de cartera debemos de restringir los pesos a exclusivamente positivos. 

- **Long-Short**: en este caso dejas que los pesos puedan ser negativos (vender en corto) pero restringes la **exposición neta** (suma total de pesos) dependiendo de si quieres Market Neutral (exposición neta o suma total cero) o Long-Biased (exposición neta positiva), el **nivel de apalancamiento** o **exposición bruta** (suma total de pesos en valor absoluto) y el **riesgo sistemático $\beta$** (suma de los pesos por las betas) para obligar al optimizador a calibrar los pesos de tal forma que la $\beta$ del lado Long cancele exactamente la $\beta$ del lado Short.

- **Exposición sectorial**: también podemos agrupar las empresas por sectores y limitar directamente la porción en un solo sector (por ejemplo, por debajo del 20%). 

- **Tracking error**: en fondos que siguen un índice se restringe lo alejado que puedes estar de ese benchmark. 

- **Factor Neutral**: se trata de reducir la exposición a un factor para poder aislar únicamente el que quieres estudiar. De esta manera especificas el factor que mueve tu cartera. 

- **Liquidez**: restricción al monto total que se destina a comprar y vender acciones en un solo día. 

- **Turnover**: limitamos el número de rebalanceos en un intervalo de tiempo o añadimos un término de penalización en la función objetivo para que el modelo no rebalancee todo el ratto. 

- **Número máximo de posiciones**: limitamos el número máximo de empresas en nuestro portfolio ya que el gestor prefiere mantener una cartera más concentrada y fácil de gestionar.

Obviamente no es necesario añadir todas las restricciones. Nos podemos quedar con una Long-Only con peso máximo por activo, matriz de covarianzas Ledoit-Wolf y penalización o restricción por turnover.

## Fase 7 — Motor de backtesting (1-1.5 semanas)

- **Costes de transacción**: aplicad un coste proporcional al turnover en cada rebalanceo (por ejemplo 5-10 puntos básicos por operación, valores típicos usados en literatura académica para acciones líquidas de gran capitalización). Sin esto, cualquier estrategia con rebalanceo frecuente parece mejor de lo que sería en la realidad.
- **Turnover**: calculadlo explícitamente en cada periodo (cuánto cambia la composición de la cartera) — es una métrica que cualquier entrevistador de un fondo sistemático os va a pedir si presentáis una estrategia.
- **Implementación**: evitad un backtest "vectorizado ingenuo" que no respete el orden temporal correcto entre señal, decisión y ejecución (es decir, si la señal se calcula con datos hasta el cierre del día X, la cartera no puede ejecutarse al precio de cierre del mismo día X, hay que asumir ejecución al día siguiente como mínimo). Este tipo de detalle de "look-ahead" en la ejecución es un error muy común y muy criticado cuando aparece en un backtest.

**División de trabajo**: tu compañero construye el motor de backtesting de forma genérica y reutilizable (que pueda aceptar cualquier señal de pesos y devolver la serie de retornos de cartera); tú defines qué supuestos de costes y ejecución son realistas para el tipo de activos que estáis usando.

---
---

#### **¿Habría ganado dinero esta estrategia si la hubiese ejecutado en el pasado?**

Una vez tenemos la estadística, el machine learning y la optimización nos toca comprobar si realmente nuestro modelo es rentable. Esto se hace con el **backtest** que sería una simulación lo más fiel posible de cómo habría funcionado la estrategia si se hubiera ejecutado en tiempo real. Podemos ver el backtest como un bucle temporal: construye señales, construye cartera, ejecuta órdenes, calcula retorno, actualiza patrimonio y vuelta a empezar. La clave aquí no es hacer el backtest sino que este sea lo más realista posible y aquí es donde entran los tres principales problemas: 

- **Costes de transacción**: comprar y vender acciones no es gratis por lo que a la rentabilidad bruta obtenida le deberíamos de descontar los costes para tener la **rentabilidad neta**. Estos costes se calculan de forma sencilla multiplicando la rotación de cartera mensual (cuánto cambia en porcentaje tu cartera mes tras mes) por el coste de una transacción, dándonos así el porcentaje de nuestro patrimonio que nos dejamos mesualmente en costes de transacción. 

- **Turnover**: el cambio en la cartera entre dos rebalanceos se puede calcular como el sumatorio de las diferencias entre pesos nuevos y pesos viejos y se utiliza para ver cómo de alto es nuestro turnover, lo cual implica más coste y menor rentabilidad. 

- **Look-Ahead Bias**: si el mercado cierra a las 16:00 el cálculo de las señales, el entrenamiento del modelo y la optimización se hace "overnight" por lo que la señal de compra en ningún caso se puede realizar a las 16:00 de ese mismo día (estarías viajando al pasado) sino a la mañana siguiente. 

la arquitectura completa del backtest quedaría: extraes los datos disponibles hasta fecha $t$, calculas las features, entrenas el modelo, optimizas la cartera, aplicas costes y restricciones, ejecutas en $t+1$, observas retornos, actualizas patrimonio y vuelves a iniciar en la fecha siguiente.

## Fase 8 — Evaluación de performance (3-5 días)

Más allá del retorno total, las métricas que de verdad se miran en la industria:

- Sharpe ratio (anualizado), y si os apetece, Sortino (que solo penaliza la volatilidad negativa, más relevante para inversores reales que el Sharpe clásico).
- Maximum drawdown y tiempo de recuperación.
- Calmar ratio (retorno anualizado / max drawdown), útil para comunicar el riesgo de cola de forma simple.
- Comparación contra benchmarks honestos: no solo "buy & hold del índice", sino también contra una cartera de pesos iguales o contra el decil superior sin optimización (para aislar cuánto valor añade específicamente la parte de ML frente a simplemente seguir el ranking de un factor simple como momentum puro).
- Análisis de subperiodos: ¿la estrategia funciona de forma consistente o todo el resultado viene de 2020-2021 (un periodo muy atípico para los mercados)? Esto es clave para detectar si el resultado es robusto o es un artefacto de un periodo concreto.

---
--- 

#### **¿Cómo sé si esta curva de patrimonio es realmente buena?**

El hecho de haber pasado el examen (superar el backtest) no significa nada por si solo ya que no se tiene en cuenta el riesgo que hemos corrido o los drawdowns que nos hemos comido. Para eso existen las **métricas de performance** que ya no evaluan cuánto gana una estrategia sino de que forma se ha ganado ese dinero. Estas métricas van a responder a cinco preguntas: 

- **¿Cuánto gana?**: retorno anualizado y CAGR.

- **¿Cuánto riesgo asume?**: Sharpe, Sortino y volatilidad. 

- **¿Cómo de malas son las crisis?**: Maximum Drawdown, Recovery Time y Calmar.

- **¿Es mejor que alternativas razonables?**: Benchmarks. 

- **¿Es robusta?**: análisis por subperiodos. 

---

**¿Mi cartera genera buena rentabilidad ajustada al riesgo?**

El Sharpe ratio es una medida bastante intuitiva de la cantidad de retorno que obtengo por cada unidad de riesgo que asumo, digamos que mide cómo de eficiente es nuestra cartera para generar rentabilidad asumiendo el mínimo riesgo posible. De alguna manera estaría capturando estabilidad ya que un Sharpe elevado, especialmente si se mantiene de forma consistente en distintos periodos y después de costes de transacción, es una evidencia mucho más sólida de que la estrategia contiene un alpha real y no solo una exposición elevada al mercado.

A pesar de sus limitaciones sigue siendo la métrica más utilizada porque es sencillo de calcular, comparable entre estrategias, independiente de la escala del capital e intuitivo para gestores e inversores. Destacar que en nuestro caso más que comparar carteras estamos comparando estrategias enteras (Top decil Equal Weight vs ML + Markowitz). 

---

**¿Las pérdidas grandes en mi cartera son frecuentes o está dominada por volatilidad positiva?**

El **Sortino ratio** justamente resuelve el problema del Sharpe ya que solamente tiene en cuenta la volatilidad hacia abajo (downside deviation) calculada únicamente con los retornos negativos (o inferiores al objetivo). Obviamente sigue teniendo problemas como que sigue siendo una medida basada en momentos (no sabe nada de crashes, heavy tails o riesgo de liquidez), ignora la forma del drawdown y depende del umbral elegido (la "mala rentabilidad" puede ser la negativa pero también la del tipo libre de riesgo o la que está por debajo del objetivo a cumplir). 

---

**¿Qué es lo peor que me puede pasar si invierto?**

Hasta ahora hablado de métricas relativas a la eficiencia (combinación de la rentabilidad y el riesgo) pero no es lo único que importa. El Maximum Drwadown te dice cuál fue la mayor caída desde un máximo histórico hasta el mínimo posterior antes de alcanzar un nuevo máximo, es decir, no es una medida de la eficiencia sino de la peor experiencia que tuvo un inversor siguiendo esta estrategia. 

Esto conecta mucho con la psicología del inversor, una estrategia no solo debe ser rentable sino invertible. Una estrategia que cae un 70% puede ser matemáticamente excelente pero psicológicamente casi imposible de seguir, por eso los gestores prestan tantísima atención al drawdown.

Conectado con el Drawdown tenemos el **Recovery Time** que sería el tiempo que tarda la cartera en volver a superar el máximo anterior después de haber entrado en drawdown. No solo importa la gravedad de la caída sino cuánto tiempo pasó hasta volver a flote. 

> Matemáticamente recuperarse de una caída es más difícil de lo que parece: si la cartera pierde un 10% necesita ganar un 11% aproximadamente pero si pierde un 50% tendrá que ganar un 100% y si pierde un 80% tendrá que ganar un 400%. 

---

**¿Y si puediésemos combinar rentabilidad y drawdown?**

En vez de preguntarse cuánto retorno obtengo por unidad de volatilidad el **Calmar Ratio** pregunta cuánto retorno obtengo por la peor caída que tengo que soportar. La expresion del Calmar es el cociente entre el CAGR y el Maximum Drawdown en valor absoluto. El Calmar es muy popular en estrategias donde el principal riesgo no es la volatilidad diaria, sino los eventos extremos pero siempre viene bien calularlo. 

Como todos los ratios tiene sus defectos como que solo utiliza el dato del maximum drawdown (depende mucho de si hubo o no una crisis excepcional), ignora el resto de la distribución (puede haber muchos drawdowns pequeños que no contabilizan) y no considera el tiempo de recuperación.

---

**¿Mi estrategia realmente añade valor frente a alternativas razonables?**

Aquí entra la **comparación contra Benchmarks**, la cual sirve para dar una referencia clara de si nuestra rentabilidad es verdaderamente buena o no (el dato aislado de un 18% anual no nos dice nada). Esta comparación nos aclara si estamos aportando valor realmente o si podríamos haber obtenido el mismo resultado con un método mucho más sencillo. 

La primera comparación que se nos puede ocurrir es el **Buy & Hold del índice de referencia**, en nuestro caso del SP500, pero no es la única. 

- **¿El mérito es del factor o del modelo?**: si nuestra estrategia toma todas las decisiones solamente en función de un factor muy potente como puede ser el Momentum el mérito recae en el factor más que en el modelo. Esto se puede ver claro si comparamos el modelo con comprar directamente el decil con mayor Momentum y ambos obtienen prácticamente el mismo resultado. 

Es decir, cada vez que queramos añadir un nuevo componente al pipeline debemos demostrar cuánto valor añade ese componente, no basta con demostrar que el sistema completo funciona. Por ejemplo, si queremos ver cuánto aporta el ML contruimos un benchmark sin ML y si queremos ver lo que aporta Markowitz construimos otro benchmark sin optimización. Los benchmarks que podemos utilizar son: 

- **Benchmark 1**: Buy & Hold del S&P 500 ó Top decil por Momentum + Equal Weight (comparas estrategia completa)

- **Benchmark 2**: Top decil por Momentum + Optimización (comparas la parte de ML)

- **Benchmark 3**: Top decil por ML + Equal Weight (comparas la parte de optimización)

> Podríamos introducir un benchmark adicional: Random Selection + Optimizador. Si tu ML + Optimizador no bate de forma estadísticamente significativa a una selección aleatoria repetida 100 veces (Monte Carlo), el ML solo está capturando ruido.

La clave es que no basta con demostrar que una estrategia funciona, hay que demostrar por qué funciona. Los benchmarks son precisamente la herramienta que permite aislar el efecto de cada decisión de diseño: si tu estrategia supera al índice, ya sabes que añade valor frente al mercado, si además supera a una cartera de igual ponderación con los mismos activos, puedes atribuir parte de la mejora a la optimización y si también supera a una estrategia basada únicamente en un factor simple como el momentum, entonces tienes evidencias de que el modelo de Machine Learning está capturando información adicional.

--- 

**¿La estrategia funciona de forma consistente en el tiempo o depende de unos pocos años extraordinarios?**

El anáisis por subperiodos intenta responder a esta pregunta y lo que se intenta ver es si realmente la rentabilidad viene de una ganancia estable año tras año o si viene de uno o dos años excepcionales que compensaron el resto de años malos. Lo más sencillo es dividir el backtest y calular las métricas como el CAGR, Sharpe o Drawdown para cada bloque Estos bloques pueden ser grupos de dos años, año a año o algo más interesante: dividir por regímenes de mercado (alcista, bajista y crisis). 

Otra manera de ver cómo evolucionan las métricas a lo largo del tiempo es a través de un **Rolling Performance** en el cual en vez de dividir en bloques fijos calulas una métrica en un intervalo concreto y vas desplazando esa ventana de tiempo. En vez de tener un único número ahora tenemos una evolución continua de la calidad de la estrategia. 

> A este respecto se suele meter el Sharpe rolling de 36 meses, el Drawdown rolling o la rentabilidad anual por años (heatmap o barras).

## Fase 9 — Robustez y análisis crítico (3-5 días)

Esta fase es la que más distingue un proyecto de nivel alto de uno mediocre, y es rápida de hacer si las fases anteriores están bien construidas:

- **Sensibilidad a hiperparámetros**: ¿el resultado cambia mucho si usáis una ventana de entrenamiento de 3 años en vez de 5? ¿Si cambiáis el horizonte de predicción de 1 a 3 meses?
- **Sensibilidad al universo**: si tenéis tiempo, probad con un subconjunto distinto de activos (por ejemplo solo un sector) para ver si la señal generaliza o es específica de la composición elegida.
- **Comparación con regímenes de mercado** (aquí podéis conectar con la opción 2 de la que hablamos si os queda tiempo): ¿el modelo funciona mejor en mercados tranquilos que en mercados de alta volatilidad?

---
---

**¿Estoy seguro de que funciona y no es simplemente una casualidad?**

Ya no queremos realizar pruebas para ver si el modelo funciona, ahora, una vez sabemos que funciona vamos a intentar romperlo (encontrar sus puntos débiles). Esta idea se basa en el hecho de que una estrategia no es buena porque tenga un buen backtest sino porque sigue funcionando cuando intentas romperla. En otras palabras queremos ver lo robusto que es nuestro modelo (que el rendimiento no cambia drásticamente) si modificamos ligeramente supuestos razonables. Esta filosofía demuestra que no estamos intentando vender algo bonito y que realmente hay un análisis de autocrítica. Las pruebas que vamos a hacer son: 

- **Sensibilidad a hiperparámetros**: si el modelo depende demasiado de cómo los hemos configurado.

- **Sensibilidad al universo**: si la señal existe realmente o solamente en los activos que hemos elegido.

- **Regímenes de mercado**: si la estrategia funciona siempre o solo bajo determinadas condiciones.
