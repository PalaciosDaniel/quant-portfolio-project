```python
# -----------------------------
# Configuration
# -----------------------------

PARQUET_PATH = "../data/raw/sp500_prices.parquet"

prices = pd.read_parquet(PARQUET_PATH)
df = prices.copy()

START_YEAR = 2010
END_YEAR = 2024


# ============================================================
# Load data
# ============================================================

def get_trading_periods_by_ticker(
    df: pd.DataFrame,
    price_field: str = "Close"
) -> dict:
    """
    Identify the first and last trading date for each ticker based on
    non-missing values in the selected price field.

    Returns two groups:
    - full_period: tickers with data from the dataset's first to last date.
    - partial_period: tickers that start later, end earlier, or have no data.
    """
    if not isinstance(df, pd.DataFrame):
        raise TypeError("df must be a pandas DataFrame.")

    if not isinstance(df.columns, pd.MultiIndex):
        raise ValueError("Columns must be a MultiIndex with Price and Ticker levels.")

    if not isinstance(df.index, pd.DatetimeIndex):
        raise ValueError("Index must be a DatetimeIndex.")

    # Identify MultiIndex levels regardless of their column order.
    price_level = "Price" if "Price" in df.columns.names else 0
    ticker_level = "Ticker" if "Ticker" in df.columns.names else 1

    available_fields = df.columns.get_level_values(price_level).unique()
    if price_field not in available_fields:
        raise ValueError(f"'{price_field}' is not available in the dataset.")

    # Define the global date range of the dataset.
    dataset_start = df.index.min()
    dataset_end = df.index.max()

    tickers = df.columns.get_level_values(ticker_level).unique()
    results = []

    for ticker in tickers:
        # Extract the selected price series for the current ticker.
        ticker_prices = df.xs(ticker, level=ticker_level, axis=1)[price_field]

        # Keep only dates with an observed price.
        valid_dates = ticker_prices.dropna().index

        if len(valid_dates) == 0:
            first_trading_date = pd.NaT
            last_trading_date = pd.NaT
            traded_full_period = False
        else:
            first_trading_date = valid_dates.min()
            last_trading_date = valid_dates.max()

            # A ticker belongs to the full-period group if it has data
            # at both boundaries of the complete dataset.
            traded_full_period = (
                first_trading_date == dataset_start
                and last_trading_date == dataset_end
            )

        results.append({
            "Ticker": ticker,
            "first_trading_date": first_trading_date,
            "last_trading_date": last_trading_date,
            "traded_full_period": traded_full_period,
        })

    periods = pd.DataFrame(results).set_index("Ticker")

    full_period = periods[periods["traded_full_period"]].sort_index()

    partial_period = (
        periods[~periods["traded_full_period"]]
        .sort_values(
        "first_trading_date",
        ascending=False,
        na_position="last",
    )
)

    return {
    "full_period": full_period,
    "partial_period": partial_period,
    }
```

```python
def compute_rolling_volatility(log_returns, window=252):
    """Calculate rolling volatility using logarithmic returns."""
    return log_returns.rolling(window=window).std()


def compute_downside_volatility(log_returns, window=252):
    """Calculate rolling volatility using only negative logarithmic returns."""
    negative_returns = log_returns.where(log_returns < 0)

    return negative_returns.rolling(
        window=window,
        min_periods=21,
    ).std()


def compute_upside_volatility(log_returns, window=252):
    """Calculate rolling volatility using only positive logarithmic returns."""
    positive_returns = log_returns.where(log_returns > 0)

    return positive_returns.rolling(
        window=window,
        min_periods=21,
    ).std()


def compute_low_volatility_factor(log_returns, window=252):
    """Calculate the low-volatility factor as negative rolling volatility."""
    rolling_volatility = compute_rolling_volatility(
        log_returns,
        window=window,
    )

    return -rolling_volatility
```

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
from matplotlib.lines import Line2D

# ============================================================
# Post 5 — Standard K-Fold vs CPCV (IG 4:5 Format)
# ============================================================

# Colors
BACKGROUND = "#F8F9FA"
TRAIN_COLOR = "#1B3B6F"
VALIDATION_COLOR = "#2A9D8F"
PURGE_COLOR = "#E05263"
EMBARGO_COLOR = "#8D99AE"
TEXT_COLOR = "#40464D"
TITLE_COLOR = "#1B3B6F"

# ============================================================
# Figure
# ============================================================

fig, axes = plt.subplots(
    2,
    1,
    figsize=(8, 10),  # Aspect ratio 4:5
    gridspec_kw={"hspace": 0.40},
)

fig.patch.set_facecolor(BACKGROUND)

# ============================================================
# Helper function
# ============================================================

def add_block(
    ax,
    start,
    width,
    color,
    label=None,
    text_color="white",
    fontsize=10,
):
    """
    Add a horizontal block to the schematic.
    """
    rectangle = Rectangle(
        (start, 0.25),
        width,
        0.5,
        facecolor=color,
        edgecolor="none",
    )
    ax.add_patch(rectangle)

    if label is not None:
        ax.text(
            start + width / 2,
            0.5,
            label,
            ha="center",
            va="center",
            fontsize=fontsize,
            fontweight="bold",
            color=text_color,
        )

# ============================================================
# Panel 1 — Standard K-Fold
# ============================================================

ax = axes[0]

add_block(ax, start=0, width=4, color=TRAIN_COLOR, label="Train")
add_block(ax, start=4, width=2, color=VALIDATION_COLOR, label="Validation")
add_block(ax, start=6, width=4, color=TRAIN_COLOR, label="Train")

ax.set_title(
    "Standard K-Fold\n(Data Leakage)",
    fontsize=16,
    fontweight="bold",
    color=TITLE_COLOR,
    pad=20,
)

# ============================================================
# Panel 2 — CPCV
# ============================================================

ax = axes[1]

add_block(ax, start=0, width=3.6, color=TRAIN_COLOR, label="Train")
add_block(ax, start=3.6, width=0.8, color=PURGE_COLOR, label="Purge", fontsize=8)
add_block(ax, start=4.4, width=2, color=VALIDATION_COLOR, label="Validation")
add_block(ax, start=6.4, width=0.8, color=EMBARGO_COLOR, label="Embargo", fontsize=8)
add_block(ax, start=7.2, width=2.8, color=TRAIN_COLOR, label="Train")

ax.set_title(
    "Combinatorial Purged Cross-Validation\n(No Leakage)",
    fontsize=16,
    fontweight="bold",
    color=TITLE_COLOR,
    pad=20,
)

# ============================================================
# Common axis formatting
# ============================================================

for ax in axes:
    ax.set_xlim(0, 10)
    ax.set_ylim(0, 1)

    # Hide numerical axes
    ax.set_xticks([])
    ax.set_yticks([])

    # Remove spines
    for spine in ax.spines.values():
        spine.set_visible(False)

    ax.set_facecolor(BACKGROUND)

# ============================================================
# General legend
# ============================================================

legend_elements = [
    Line2D([0], [0], color=TRAIN_COLOR, linewidth=10, label="Train"),
    Line2D([0], [0], color=VALIDATION_COLOR, linewidth=10, label="Validation"),
    Line2D([0], [0], color=PURGE_COLOR, linewidth=10, label="Purge"),
    Line2D([0], [0], color=EMBARGO_COLOR, linewidth=10, label="Embargo"),
]

fig.legend(
    handles=legend_elements,
    loc="upper center",
    bbox_to_anchor=(0.5, 0.56),
    ncol=2,  # Cambiado a 2 columnas para el formato vertical
    frameon=False,
    fontsize=11,
)

# ============================================================
# Methodology information strip
# ============================================================

fig.text(
    0.5,
    0.86,
    "CPCV configuration",
    ha="center",
    va="center",
    fontsize=12,
    fontweight="bold",
    color=TITLE_COLOR,
)

fig.text(
    0.5,
    0.82,
    "7 blocks  •  21 folds  •  5 Train / 2 Validation\nPurge: 21 business days  •  Embargo: 11 business days",
    ha="center",
    va="center",
    fontsize=10.5,
    color=TEXT_COLOR,
    linespacing=1.6,
)

# ============================================================
# Small explanatory note
# ============================================================

fig.text(
    0.5,
    0.05,
    "7 blocks → choose 2 for validation → C(7,2) = 21 combinations",
    ha="center",
    va="center",
    fontsize=10,
    color=TEXT_COLOR,
)

# ============================================================
# Layout
# ============================================================

plt.tight_layout(
    rect=[0, 0.08, 1, 0.77]
)

# ============================================================
# Export
# ============================================================

plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# ============================================================
# Model Performance Table (Vertical Format)
# Estilo basado en Factor Screening Table
# ============================================================

# ============================================================
# Colors & Style Settings
# ============================================================

BACKGROUND = "#F8F9FA"

TITLE_COLOR = "#1B3B6F"
TEXT_COLOR = "#40464D"
SUBTITLE_COLOR = "#59636E"

SEPARATOR_COLOR = "#D0D4D8"

# Color para la cabecera de modelos (usamos el azul principal)
MODEL_HEADER_BG = "#1B3B6F" 

LABEL_COL_BG = "#1B3B6F"
LABEL_COL_TEXT = "#FFFFFF"

ROW_BG_EVEN = "#FFFFFF"
ROW_BG_ODD = "#EFF1F3"

HEADER_TEXT = "#FFFFFF"

NUMBER_COLOR = "#1A1A1A"

# ============================================================
# Data Processing (Transposed for Vertical Layout)
# ============================================================

# Tus datos originales organizados por modelo
raw_data = {
    "Metric": [
        "Rank IC Mean",
        "Rank IC Median",
        "Rank IC Std",
        "Rank IC Information Ratio",
        "Rank IC Hit Rate (%)",
        "Rank IC t-statistic",
        "RMSE",
        "MAE"
    ],
    "Ridge": [0.0522, 0.0582, 0.2304, 0.2264, 59.6, 84.37, 0.1061, 0.0748],
    "XGBoost": [0.0550, 0.0573, 0.2035, 0.2704, 60.4, 85.22, 0.1061, 0.0749],
    "Random Forest": [0.0546, 0.0612, 0.2124, 0.2573, 60.4, 84.96, 0.1060, 0.0750]
}

# Creamos DataFrame para formatear números fácilmente
df = pd.DataFrame(raw_data)

# Formateo: 4 decimales para IC y errores, 1 para t-stat y %, 2 para t-stat largo
def format_values(row):
    metric = row['Metric']
    if '%' in metric or 't-statistic' in metric:
        fmt = "{:.1f}" if 't-statistic' not in metric else "{:.2f}"
        return [fmt.format(row['Ridge']), fmt.format(row['XGBoost']), fmt.format(row['Random Forest'])]
    else:
        fmt = "{:.4f}"
        return [fmt.format(row['Ridge']), fmt.format(row['XGBoost']), fmt.format(row['Random Forest'])]

# Aplicar formato y convertir a lista de listas para Matplotlib
formatted_body = []
for _, row in df.iterrows():
    # Helper para romper textos largos de métricas y que quepan mejor
    pretty_metric = row['Metric'].replace("Rank IC ", "Rank IC\n").replace("Information Ratio", "Info.\nRatio")
    formatted_body.append([pretty_metric] + format_values(row))

# Cabeceras (Modelos)
model_headers = ["Ridge", "XGBoost", "Random\nForest"]
col_labels = ["Metric"] + model_headers

n_models = len(model_headers)
n_metrics = len(formatted_body)

# ============================================================
# Figure Setup (4:5 Aspect Ratio for LinkedIn/Insta)
# ============================================================

fig = plt.figure(figsize=(9, 11.25)) # 1080x1350 roughly at 120 dpi
fig.patch.set_facecolor(BACKGROUND)

# Ax auxiliar para textos y layout general
ax0 = fig.add_axes([0, 0, 1, 1])
ax0.set_facecolor("none")
ax0.set_xlim(0, 1)
ax0.set_ylim(0, 1)
ax0.axis("off")

# ------------------------------------------------------------
# Header Titles
# ------------------------------------------------------------

ax0.text(
    0.5, 0.965, "MODEL COMPARISON \u00b7 CROSS-VALIDATION",
    transform=ax0.transAxes,
    ha="center", va="center",
    fontsize=11.5, fontweight="bold", color=TITLE_COLOR,
)

ax0.text(
    0.5, 0.920, "Performance Metrics",
    transform=ax0.transAxes,
    ha="center", va="center",
    fontsize=24, fontweight="bold", color=TITLE_COLOR,
)

ax0.text(
    0.5, 0.885, "Comparison of information coefficient and error metrics",
    transform=ax0.transAxes,
    ha="center", va="center",
    fontsize=12, color=SUBTITLE_COLOR,
)

# ------------------------------------------------------------
# Table Axes Setup
# ------------------------------------------------------------

# Ajustamos la posición para formato vertical
ax1 = fig.add_axes([0.05, 0.25, 0.90, 0.60])
ax1.axis("off")

table = ax1.table(
    cellText=formatted_body,
    colLabels=col_labels,
    cellLoc="center",
    loc="center",
)

table.auto_set_font_size(False)
table.set_fontsize(12)
# Escalado vertical para dar aire a las filas
table.scale(1, 3.8) 

# Ajuste de anchos de columna
n_cols = len(col_labels)
label_col_width = 0.30 # Más ancho para las descripciones de métricas
model_col_width = (1 - label_col_width) / n_models
col_widths = [label_col_width] + [model_col_width] * n_models

for row_idx in range(n_metrics + 1):
    for col_idx, width in enumerate(col_widths):
        table[(row_idx, col_idx)].set_width(width)

# ------------------------------------------------------------
# Table Styling
# ------------------------------------------------------------

# Cabecera (Fila 0)
for col_idx in range(n_cols):
    cell = table[(0, col_idx)]
    cell.set_facecolor(MODEL_HEADER_BG)
    cell.get_text().set_color(HEADER_TEXT)
    cell.get_text().set_fontweight("bold")
    cell.get_text().set_fontsize(13)
    cell.set_edgecolor(BACKGROUND)
    cell.set_linewidth(2)
    # Altura extra para la cabecera
    cell.set_height(cell.get_height() * 1.3)

# Alineado a la izquierda específico para la primera columna de la cabecera
table[(0, 0)].get_text().set_ha("left")


# Cuerpo de la tabla
for row_idx in range(1, n_metrics + 1):
    # Fondo cebra
    row_bg = ROW_BG_EVEN if row_idx % 2 == 1 else ROW_BG_ODD
    
    for col_idx in range(n_cols):
        cell = table[(row_idx, col_idx)]
        cell.set_edgecolor(SEPARATOR_COLOR)
        cell.set_linewidth(0.8)
        
        if col_idx == 0:
            # Columna de etiquetas (Métricas)
            cell.set_facecolor(row_bg) # Mismo fondo que la fila
            cell.get_text().set_color(TITLE_COLOR) # Color de texto de títulos
            cell.get_text().set_fontweight("semibold")
            cell.get_text().set_fontsize(11.5)
            cell.get_text().set_ha("left") # Alineado a la izquierda
        else:
            # Columnas de datos (Números)
            cell.set_facecolor(row_bg)
            cell.get_text().set_color(NUMBER_COLOR)
            cell.get_text().set_fontweight("normal")
            cell.get_text().set_fontsize(13)

# ------------------------------------------------------------
# Methodology Notes / Key Takeaways
# ------------------------------------------------------------

ax0.text(
    0.08, 0.210, "ANALYSIS NOTES",
    transform=ax0.transAxes, ha="left", va="center",
    fontsize=13, fontweight="bold", color=TITLE_COLOR,
)

analysis_text = (
    "\u2022 XGBoost shows the highest Information Ratio and t-statistic.\n"
    "\u2022 Random Forest achieves the highest median Rank IC.\n"
    "\u2022 Prediction errors (RMSE/MAE) are virtually identical across models.\n"
    "\u2022 All models demonstrate significant predictive power (t-stat > 84)."
)

ax0.text(
    0.08, 0.135,
    analysis_text,
    transform=ax0.transAxes, ha="left", va="center",
    fontsize=11.5, color=SUBTITLE_COLOR, linespacing=1.6,
)


# ------------------------------------------------------------
# Footer
# ------------------------------------------------------------

ax0.plot(
    [0.08, 0.92], [0.05, 0.05],
    transform=ax0.transAxes,
    color=SEPARATOR_COLOR, linewidth=0.8,
)

ax0.text(
    0.5, 0.025,
    "Rank IC \u00b7 Spearman correlation between predicted and actual return ranks",
    transform=ax0.transAxes,
    ha="center", va="center",
    fontsize=9.5, color=SUBTITLE_COLOR, style="italic",
)

# ============================================================
# Export — 1080x1350 px (LinkedIn 4:5 mobile format)
# ============================================================

# plt.savefig(
#     "model_performance_table_vertical.png",
#     dpi=120,
#     facecolor=fig.get_facecolor(),
#     bbox_inches='tight', # A veces ayuda a que no se corte nada
#     pad_inches=0.1
# )

plt.show()
```
